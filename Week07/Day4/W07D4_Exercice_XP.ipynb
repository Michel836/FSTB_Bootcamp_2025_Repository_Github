🌟 Exercice 1 : Analyse comparative de l’IA générative et de l’IA traditionnelle
L'objectif est d'évaluer de manière critique les forces et les faiblesses de l'IA générative par rapport à l'IA traditionnelle dans des scénarios complexes. Pour
chacun des scénarios réels suivants, déterminez lequel des deux serait le plus adapté. Fournissez une justification détaillée, prenant en compte des facteurs tels que la précision, la créativité, l'efficacité, les préoccupations éthiques et les coûts de calcul .

1. Diagnostic médical automatisé

Un hôpital souhaite mettre en œuvre un système d'IA pour détecter le cancer du poumon à partir d'images radiologiques. Doit-il utiliser un modèle de classification d'IA traditionnel ou une approche d'IA générative capable de synthétiser de nouvelles images médicales pour la formation ?
2. Génération de documents juridiques

Un cabinet d'avocats souhaite un système d'IA capable de rédiger des contrats juridiquement contraignants avec une intervention humaine minimale. Doit-il utiliser un système basé sur des règles (IA traditionnelle) ou un modèle de langage génératif ? Quels sont les risques encourus ?
3. Recherche scientifique générée par l'IA

Une équipe de recherche universitaire souhaite automatiser les revues de littérature en résumant des milliers d'articles universitaires et en suggérant de nouvelles hypothèses de recherche. L'IA générative serait-elle la meilleure approche ? Pourquoi ?
4. Prévisions des marchés financiers

Un fonds spéculatif utilise l'IA pour prédire les tendances boursières en s'appuyant sur des données de trading historiques. L'IA générative serait-elle utile, ou devrait-il s'appuyer sur des modèles d'IA traditionnels comme la régression et les prévisions de séries chronologiques ?
5. Prise de décision des véhicules autonomes

Les voitures autonomes doivent prendre des décisions de conduite en temps réel dans des environnements imprévisibles. L'IA générative pourrait-elle être utile, ou faut-il privilégier les modèles d'IA traditionnels ? Il faut tenir compte de la sécurité et de la fiabilité.

## 🌟 Exercice 1 : Analyse comparative de l’IA générative et de l’IA traditionnelle

## Objectif
Comparer IA générative et IA traditionnelle sur la précision, créativité, efficacité, éthique et coûts dans des cas concrets.

---

## 1. Diagnostic médical automatisé
**Choix :** IA traditionnelle (classification supervisée type CNN).

**Arguments :**
- **Précision :** les modèles traditionnels sont entraînés sur des données réelles, validées médicalement.
- **Créativité :** inutile, un modèle doit identifier des pathologies connues, pas inventer des anomalies.
- **Efficacité :** rapide et optimisé pour des tâches spécifiques de détection.
- **Éthique :** générer des images médicales artificielles pourrait introduire des erreurs médicales graves.
- **Coûts :** IA traditionnelle nécessite moins de puissance de calcul (pas de génération d'images complexes).
- **Interprétabilité :** IA traditionnelle peut être plus facilement auditable par les médecins (via heatmaps par exemple).

**Conclusion :** **IA traditionnelle** reste la référence pour la fiabilité et la sécurité dans les diagnostics médicaux.

---

## 2. Génération de documents juridiques
**Choix :** IA générative (modèle de langage avancé).

**Arguments :**
- **Précision :** IA générative peut réutiliser de nombreux modèles contractuels et les adapter.
- **Créativité :** utile pour combiner des clauses complexes adaptées à divers scénarios.
- **Efficacité :** forte automatisation de la rédaction avec gain de temps significatif.
- **Éthique :** supervision juridique impérative pour éviter des erreurs graves ou clauses abusives.
- **Coûts :** investissement initial élevé mais amorti par la productivité sur le long terme.
- **Flexibilité :** IA générative s’adapte à différents types de contrats sans reprogrammer des règles.

 **Conclusion :** **IA générative** est adaptée pour la rédaction avec un contrôle humain critique pour éviter les erreurs juridiques.

---

## 3. Recherche scientifique générée par l'IA
**Choix :** IA générative.

**Arguments :**
- **Précision :** synthèse fidèle d'informations à partir de vastes bases de données scientifiques.
- **Créativité :** propose des combinaisons d'idées nouvelles à partir de connaissances existantes.
- **Efficacité :** réduit considérablement le temps de lecture et de synthèse pour les chercheurs.
- **Éthique :** nécessite un contrôle pour éviter les résumés incorrects ou la propagation d'erreurs.
- **Coûts :** coûts raisonnables via des modèles open-source ou pré-entraînés.
- **Exploration :** permet l'identification d'angles de recherche non envisagés par des humains.

 **Conclusion :** **IA générative** est idéale pour accélérer la veille scientifique et stimuler de nouvelles idées.

---

## 4. Prévisions des marchés financiers
**Choix :** IA traditionnelle (régression, séries temporelles, modèles prédictifs explicables).

**Arguments :**
- **Précision :** forte sur des données historiques avec des techniques éprouvées.
- **Créativité :** inutile voire risquée, les marchés financiers exigent de la rigueur.
- **Efficacité :** rapide, modèles simples à entraîner et déployer.
- **Éthique :** IA générative risque d'amplifier des biais de marché ou de créer des stratégies spéculatives toxiques.
- **Coûts :** plus faible, rapide à mettre en production.
- **Régulation :** les autorités financières privilégient des systèmes explicables, plus faciles à auditer.

 **Conclusion :** **IA traditionnelle** reste la meilleure option pour les prédictions financières fiables et conformes.

---

## 5. Prise de décision des véhicules autonomes
**Choix :** IA traditionnelle.

**Arguments :**
- **Précision :** prioritaire pour éviter les accidents, IA traditionnelle est testée et validée en environnement critique.
- **Créativité :** non souhaitable car peut conduire à des comportements imprévus et dangereux.
- **Efficacité :** temps de réaction optimisé avec des modèles déterministes.
- **Éthique :** obligation légale de sécurité, IA générative pose des problèmes de validation réglementaire.
- **Coûts :** IA traditionnelle demande moins de ressources embarquées (CPU/GPU).
- **Robustesse :** IA traditionnelle est plus robuste face aux situations critiques et facilement testable via simulations.

 **Conclusion :** **IA traditionnelle** est indispensable pour garantir sécurité, fiabilité et conformité réglementaire dans les véhicules autonomes.

## 🌟 Exercice 3 : Optimisation et mise au point des modèles d’IA générative

## 1. Ingénierie rapide : Réécriture des invites

| Invite de base | Nouvelle version améliorée |
|-----------------|---------------------------|
| Générer une image d’une ville futuriste. | Générer une image ultra-détaillée d'une ville futuriste en 2150, avec des gratte-ciels organiques, des véhicules volants, un ciel nocturne néon, et une ambiance cyberpunk. |

---

## 2. Biais et équité dans les données d'entraînement

**Propositions :**
- **Diversification des sources :** collecter des données provenant de médias aux opinions variées (différents pays, idéologies).
- **Filtrage automatique :** utiliser des outils d’analyse linguistique pour détecter et équilibrer les biais avant entraînement.
- **Annotation humaine :** faire appel à des experts pour annoter et équilibrer les données sensibles.
- **Rééchantillonnage équilibré :** ajuster les proportions de données pour limiter la sur-représentation d'une opinion.
- **Audit de biais régulier :** évaluer et corriger périodiquement le contenu produit pour garantir la neutralité.

---

## 3. Ajustement précis pour tâches médicales

**Étapes clés :**
1. **Collecte et prétraitement :** réunir des publications médicales validées (PubMed, études cliniques), nettoyage des données, anonymisation.
2. **Apprentissage par transfert :** partir d’un grand modèle pré-entraîné (type LLM généraliste) puis le fine-tuner sur le corpus médical spécifique.
3. **Techniques spécifiques :** utilisation de Low-Rank Adaptation (LoRA) ou P-Tuning pour un ajustement efficace.
4. **Évaluation :**
   - Métriques spécifiques (exactitude factuelle, cohérence contextuelle).
   - Revue humaine par experts médicaux.
   - Détection automatique de "hallucinations" factuelles.
5. **Mises à jour régulières :** recyclage des données pour suivre l’évolution scientifique.

---

## 4. Indicateurs quantitatifs d’évaluation

| Type de contenu | Indicateurs Objectifs |
|-----------------|-----------------------|
| **Texte généré** | - **BLEU** : qualité des textes traduits ou générés.<br>- **ROUGE** : pertinence des résumés générés.<br>- **Perplexité** : mesure de la cohérence linguistique.<br>- **Exactitude factuelle** : évaluation via benchmarks spécialisés. |
| **Image générée** | - **FID (Fréchet Inception Distance)** : réalisme et qualité visuelle.<br>- **IS (Inception Score)** : diversité et clarté des images.<br>- **CLIP score** : correspondance texte-image. |

---

## 5. Contrôler créativité et cohérence

**Techniques :**
- **Température :**  
  - Basse température (~0.2) = génération cohérente et logique.  
  - Haute température (~0.9) = plus de créativité mais risque d'incohérence.
- **Apprentissage par renforcement (RLHF) :** guide le modèle via feedback humain pour privilégier la cohérence et réduire les erreurs absurdes.
- **Mécanismes d’attention :** focus sur les parties pertinentes du contexte, améliorant la logique et la cohérence du discours.
- **Top-k / Top-p (nucleus sampling) :** filtrage des sorties improbables tout en laissant une marge contrôlée pour la créativité.

 **Résumé :**
- **GAN** : meilleure qualité visuelle, mais instabilité et coût élevé.
- **VAE** : meilleure interprétabilité, plus stable, moins coûteux, mais qualité souvent inférieure.
- **Utilisation hybride** possible dans certains cas pratiques.


## 🌟 Exercice 5 : Exploration avancée de l'espace latent dans les VAE

## 1. Visualisation des distributions d'espace latent

**Évaluation :**
- **Vérifier la séparation des classes** en projetant les vecteurs latents via :
  - **t-SNE** ou **UMAP** : techniques de réduction de dimension non-linéaires pour visualiser la séparation des classes sur 2D.
  - **PCA** (si l’espace est raisonnablement linéaire) pour une première analyse rapide.

**Techniques de clustering :**
- **K-Means** : détecter si les clusters correspondent aux classes de chiffres (0-9).
- **DBSCAN** : identifier des regroupements sans pré-spécifier le nombre de classes.
- Vérification manuelle : colorier les points latents par label de chiffre connu (MNIST) pour évaluer la séparation visuelle.

---

## 2. Interpolation entre deux échantillons

**Méthode :**
- Encoder les deux images ("3" et "8") pour obtenir leurs vecteurs latents `z1` et `z2`.
- Interpoler entre `z1` et `z2` via interpolation linéaire :  
  `z = (1 - α) * z1 + α * z2` où `α` varie de 0 à 1.
- Décoder chaque vecteur intermédiaire pour obtenir des images de transition.

**Pourquoi c'est possible avec VAE :**
- **Espace latent continu et régulierisé** via la contrainte KL : les VAE forcent l'organisation latente (souvent normalisée).
- **GAN** : espace latent moins structuré, sans contrainte explicite -> interpolation produit souvent des résultats incohérents.

---

## 3. Contrôle de la variabilité via le terme KL

**Effet du KL-divergence :**
- **Augmenter la pondération KL :** force le modèle à rester proche de la distribution normale -> diversité réduite mais échantillons plus réguliers.
- **Réduire la pondération KL :** plus grande liberté de génération -> diversité plus forte mais qualité (fidélité) potentiellement dégradée.

**Application pratique :**
- **Synthèse vocale** : nécessité d’équilibrer naturalité (diversité) et intelligibilité (qualité).
- Exemple : un générateur vocal doit varier légèrement les intonations tout en restant compréhensible.

---

## 4. Démêler les représentations latentes

**Méthode :**
- Identifier manuellement (ou via apprentissage supervisé léger) quelle dimension latente contrôle chaque attribut (ex. couleur de cheveux).
- **Modifier uniquement la dimension associée** (ex. augmenter `z_i` correspondant à “cheveux blonds”) tout en gardant les autres constantes.
- Exemple : `z_modifié = z + λ * e_i` où `e_i` est la base unitaire sur la dimension visée.

**Pourquoi :**
- **Démêlage latent** : l’objectif du VAE est d’apprendre une séparation claire entre les facteurs de variation (ex. cheveux, âge, expression).

---

## 5. Comparaison PCA vs VAE

| Critère | PCA | VAE |
|----------|-----|-----|
| **Linéarité** | Linéaire : projections linéaires simples | Non-linéaire : utilise des réseaux neuronaux |
| **Interprétabilité** | Composantes directement interprétables via variances expliquées | Composantes latentes moins interprétables sans exploration ciblée |
| **Qualité de reconstruction** | Bonne pour données linéaires mais limitée | Meilleure pour données complexes non-linéaires |
| **Modélisation générative** | Non génératif : pas de génération d’échantillons | Génératif : peut produire de nouveaux exemples par échantillonnage latent |

---

**Résumé :** VAE offre un espace latent structuré et génératif, meilleur pour interpolation, contrôle des attributs et diversité contrôlée. PCA reste utile pour analyse rapide et interprétable.


## 🌟 Bilan

## Exercice 1 : IA générative vs IA traditionnelle
- **IA traditionnelle**  fiable, précise, efficace dans les tâches critiques (médical, finance, véhicules).
- **IA générative** utile pour tâches créatives, synthèses massives, rédaction juridique, mais nécessite **contrôle humain**.
- **Clé :** choisir en fonction du besoin : **précision stable** vs **créativité et adaptabilité**.

## Exercice 2 : Risques éthiques de l’IA générative
- Risques majeurs : **désinformation**, **fraudes**, **cyberguerre**, **malwares**, **vol de propriété intellectuelle**.
- **Solutions :**
   - **Légales :** lois, régulations internationales.
   - **Techniques :** détection automatique, transparence des modèles, contrôles humains.

## Exercice 3 : Optimisation des modèles génératifs
- **Prompt engineering :** améliore fortement la qualité.
- **Curation des données :** clé pour éviter les biais.
- **Fine-tuning** : indispensable pour usages spécialisés (médical, juridique).
- **Évaluation :** utiliser des métriques objectives (BLEU, FID, Perplexité).
- **Contrôle créativité/cohérence :** ajuster température, KL, RLHF selon objectifs.

## Exercice 4 : GAN vs VAE
- **GAN** : meilleure qualité visuelle (images haute résolution), utile pour génération réaliste.
- **VAE** : meilleur **contrôle latent**, **stabilité**, **interpolation fluide**.
- **Choix dépend** : GAN = qualité brute ; VAE = contrôle, apprentissage structuré.
- **Cas hybrides** possibles (VAE-GAN).

## Exercice 5 : Espace latent des VAE
- **Espace latent structuré** permet : 
   - **visualisation** (t-SNE, UMAP),
   - **interpolation continue**,
   - **contrôle ciblé des attributs**,
   - **régulation diversité vs qualité via KL**.
- **VAE** = modèle **contrôlable et génératif**, utile pour recherche scientifique, arts contrôlés, détection anomalies.

---

##  Synthèse globale :
- **IA traditionnelle** = robustesse, explicabilité.
- **IA générative** = créativité, synthèse massive, mais **risques éthiques réels**.
- **Optimisation IA** passe par : **données bien choisies**, **contrôle des prompts**, **ajustements précis**, **métriques claires**.
- **GAN** = visuel haut de gamme, **VAE** = contrôle et cohérence.
