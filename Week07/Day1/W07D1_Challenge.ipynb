## 1. Comprendre BERT et XLM-RoBERTa

````python
# ==============================================
# INSTALLATION DES DEPENDANCES
# ==============================================
!pip install --quiet transformers sentencepiece

# ==============================================
# IMPORTATION DES LIBRAIRIES
# ==============================================
from transformers import BertTokenizer, XLMRobertaTokenizer

# ==============================================
# INITIALISATION DES TOKENIZERS
# ==============================================
# BERT : modèle anglais
bert_tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")

# XLM-RoBERTa : modèle multilingue (100 langues)
xlm_tokenizer = XLMRobertaTokenizer.from_pretrained("xlm-roberta-base")

# ==============================================
# TOKENISATION DE TEXTES D'EXEMPLE
# ==============================================
texte_anglais = "Hello, how are you?"
tokens_bert = bert_tokenizer.tokenize(texte_anglais)

texte_francais = "Bonjour, comment ça va ?"
tokens_xlm = xlm_tokenizer.tokenize(texte_francais)

# ==============================================
# AFFICHAGE
# ==============================================
print("Tokenisation BERT :", tokens_bert)
print("Tokenisation XLM-RoBERTa :", tokens_xlm)

# ==============================================
# EXPLICATION :
# BERT : WordPiece (préfixe '##'), optimisé anglais.
# XLM-RoBERTa : BPE (préfixe '▁'), multilingue (100 langues).
````

Analyse rapide :

* **Résultats de tokenisation** : corrects.

  * BERT découpe mot à mot sans sous-mots car la phrase est simple.
  * XLM-RoBERTa utilise les préfixes `▁`, typique du BPE multilingue, normal.

## 2. Tokenisation du texte

````python
# ==============================================
# INSTALLATION DES DEPENDANCES
# ==============================================
!pip install --quiet transformers sentencepiece

# ==============================================
# IMPORTATION DES LIBRAIRIES
# ==============================================
from transformers import BertTokenizer, XLMRobertaTokenizer

# ==============================================
# INITIALISATION DES TOKENIZERS
# ==============================================
bert_tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
xlm_tokenizer = XLMRobertaTokenizer.from_pretrained("xlm-roberta-base")

# ==============================================
# EXEMPLE : UNE PHRASE (BERT)
# ==============================================
texte = "Hello, how are you?"
encoding_bert = bert_tokenizer.encode_plus(
    texte,
    truncation=True,
    padding='max_length',
    max_length=10,
    return_tensors='pt'
)

print("BERT input_ids :", encoding_bert['input_ids'])
print("BERT attention_mask :", encoding_bert['attention_mask'])
print("Décodage BERT :", bert_tokenizer.decode(encoding_bert['input_ids'][0]))

# ==============================================
# EXEMPLE : DEUX PHRASES (XLM-RoBERTa)
# ==============================================
texte1 = "Bonjour, comment ça va ?"
texte2 = "Très bien merci."

encoding_xlm = xlm_tokenizer.encode_plus(
    texte1,
    texte2,
    truncation=True,
    padding='max_length',
    max_length=15,
    return_tensors='pt'
)

print("XLM input_ids :", encoding_xlm['input_ids'])
print("XLM attention_mask :", encoding_xlm['attention_mask'])
print("Décodage XLM :", xlm_tokenizer.decode(encoding_xlm['input_ids'][0]))

# ==============================================
# EXPLICATIONS RAPIDES :
# ==============================================
# input_ids : ID numérique de chaque token.
# attention_mask : 1 = token utile, 0 = padding.
# labels : utilisés uniquement pour l'entraînement (pas montré ici).
# encode_plus() : produit tous ces éléments en une seule étape.
# decode() : permet de revenir des ID aux mots originaux.
````

Analyse rapide :

###  **BERT**

* `input_ids` : OK → ajout des tokens spéciaux `[CLS]` (101), `[SEP]` (102), et padding (0).
* `attention_mask` : 1 = vrais tokens, 0 = padding (correct).
* `decode()` : bonne reconstruction, **les `[PAD]` sont visibles**, mais non pris en compte par `attention_mask` → normal.

###  **XLM-RoBERTa**

* `input_ids` : OK → `<s>` (0), `</s>` (2), `<pad>` (1), avec séparation correcte des phrases.
* `attention_mask` : bonne gestion du padding (1 pour tokens valides, 0 pour padding).
* `decode()` : les balises `<s>` et `</s>` sont visibles, **normal**, car XLM-RoBERTa conserve explicitement ces tokens.

---

###  **Conclusion** :

* Les **tokens spéciaux sont bien intégrés**.
* L'**attention\_mask est correct**, prêt pour entraînement.
* **Décodage fonctionne** malgré affichage visible des tokens spéciaux — normal avec `decode()` sans nettoyage supplémentaire.

**Tout est fonctionnel**.


## 3. Préparation des données d'entrée pour le modèle

````python
# ==============================================
# INSTALLATION DES DEPENDANCES
# ==============================================
!pip install --quiet transformers sentencepiece

# ==============================================
# IMPORTATION
# ==============================================
from transformers import BertTokenizer, XLMRobertaTokenizer

# ==============================================
# INITIALISATION
# ==============================================
bert_tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
xlm_tokenizer = XLMRobertaTokenizer.from_pretrained("xlm-roberta-base")

# ==============================================
# TOKENISATION AVEC PADDING ET TRONQUAGE
# ==============================================
texte = "Ceci est un exemple de phrase pour la démonstration."

encoding = xlm_tokenizer.encode_plus(
    texte,
    truncation=True,              # coupe si trop long
    padding='max_length',         # complète jusqu'à max_length
    max_length=12,                # longueur cible
    return_attention_mask=True,
    return_tensors='pt'           # tensors PyTorch
)

print("input_ids :", encoding['input_ids'])
print("attention_mask :", encoding['attention_mask'])

# ==============================================
# JETONS SPÉCIAUX
# ==============================================
print("Jetons spéciaux (XLM-RoBERTa) :", xlm_tokenizer.special_tokens_map)

# Exemples typiques :
# <s> = début de séquence
# </s> = fin de séquence
# <pad> = remplissage (padding)
# <unk> = mot inconnu
# <mask> = pour le masquage lors de l'entraînement

# ==============================================
# TAILLE DU VOCABULAIRE
# ==============================================
print("Taille du vocabulaire XLM-RoBERTa :", xlm_tokenizer.vocab_size)

# ==============================================
# EXPLICATION RAPIDE
# ==============================================
# - input_ids = tokens convertis en IDs numériques
# - attention_mask = 1 pour tokens réels, 0 pour padding
# - special_tokens_map = dictionnaire des jetons spéciaux
# - vocab_size = taille totale du vocabulaire connu par le modèle
````

Analyse rapide :

* **input\_ids** : séquence encodée avec `<s>` (0) au début et `</s>` (2) à la fin → **normal** pour XLM-RoBERTa.

* **attention\_mask** : uniquement des 1 → aucune troncature ni padding (12 tokens, donc inférieur au `max_length` probable).

* **special\_tokens\_map** : cohérent pour XLM-RoBERTa :

    * `<s>` et `</s>` servent de début/fin.
    * `<pad>` pour le padding.
    * `<mask>` pour le masquage MLM.
    * `<unk>` pour tokens inconnus.

* **vocab\_size** : 250002, valeur correcte pour `xlm-roberta-base`.

* **Conclusion** :
Tout est **conforme**, séquences prêtes pour le modèle, vocabulaire complet et tokenisation bien encadrée par les tokens spéciaux.

## 5. Création de plis de validation croisée

````python
# ==============================================
# INSTALLATION DES LIBRAIRIES
# ==============================================
!pip install --quiet pandas scikit-learn

# ==============================================
# IMPORTATIONS
# ==============================================
import pandas as pd
from sklearn.model_selection import StratifiedKFold

# ==============================================
# CHARGEMENT D'UN EXEMPLE DE DATASET
# ==============================================
# Exemple rapide : dataframe fictif
data = {'text': [f"phrase {i}" for i in range(20)],
        'label': [0, 1]*10}
df = pd.DataFrame(data)

print("Structure du dataset :", df.shape)
print(df.head())

# ==============================================
# STRATIFIED KFOLD CROSS VALIDATION
# ==============================================
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

train_indices = []
val_indices = []

for fold, (train_idx, val_idx) in enumerate(kf.split(df, df['label'])):
    print(f"\nPli {fold+1}:")
    print("Train indices :", train_idx)
    print("Validation indices :", val_idx)
    
    train_indices.append(train_idx)
    val_indices.append(val_idx)

# ==============================================
# EXPLICATION :
# - StratifiedKFold : équilibre des étiquettes dans chaque pli.
# - shuffle=True : mélange avant découpage.
# - train_indices et val_indices : contiennent les indices pour l'entraînement et la validation.
````

Analyse rapide :

 **Dataset** :

* 20 lignes, 2 colonnes (`text`, `label`), échantillon de test pour démonstration.

 **Validation croisée stratifiée (5 plis)** :

* Chaque pli contient **4 indices validation** (20 ÷ 5 = 4), et 16 indices entraînement.
* La séparation est correcte : les indices validation changent à chaque pli.
* Comme les labels sont équilibrés (0,1 alternés), **StratifiedKFold** a bien respecté la proportion des classes dans chaque pli.

 **Conclusion** :
StratifiedKFold fonctionne correctement : bonne répartition, pas de fuite de données, préparation idéale pour entraînement + validation fiables.
