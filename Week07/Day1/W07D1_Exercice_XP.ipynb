## Exercice 1 : PNL traditionnelle et moderne : une analyse comparative

### 1. Tableau comparatif

| **Aspect**                         | **PNL Traditionnelle**                                                                              | **PNL Moderne**                                                                                  |
| ---------------------------------- | --------------------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------------ |
| **Ingénierie des fonctionnalités** | **Manuelle** : extraction manuelle de caractéristiques (TF-IDF, n-grammes)                          | **Automatique** : représentations apprises directement par le modèle                             |
| **Représentations de mots**        | **Statiques** : embeddings fixes (Word2Vec, GloVe) ou vecteurs simples                              | **Contextuelles** : embeddings dynamiques (BERT, GPT) qui dépendent du contexte                  |
| **Architectures de modèles**       | **Superficielles** : modèles simples, souvent linéaires ou peu profonds (Naïve Bayes, SVM)          | **Profondes** : réseaux neuronaux profonds (transformers, LSTM, BERT)                            |
| **Méthodologie de formation**      | **Spécifique à la tâche** : entraînement sur chaque tâche spécifique                                | **Pré-formation + Fine-tuning** : pré-formation sur des corpus massifs puis adaptation fine      |
| **Exemples clés de modèles**       | Naïve Bayes, SVM, HMM, TF-IDF                                                                       | BERT, GPT, RoBERTa, T5, Llama                                                                    |
| **Avantages**                      | - Simplicité<br>- Peu coûteux en calcul<br>- Interprétable                                          | - Performances nettement supérieures<br>- Adaptabilité<br>- Compréhension contextuelle           |
| **Inconvénients**                  | - Limité en complexité linguistique<br>- Faible généralisation<br>- Nécessite beaucoup d’ingénierie | - Coûts de calcul très élevés<br>- Moins interprétable<br>- Dépendance à des ressources massives |

---

### 2. Impact sur l’évolutivité et l’efficacité

Le passage de la PNL traditionnelle à la PNL moderne a profondément amélioré **l’évolutivité** en permettant de traiter des tâches variées via des modèles pré-entraînés, sans repartir de zéro à chaque tâche. Les modèles modernes sont **plus efficaces** en termes de performance (précision, pertinence des résultats) mais **plus coûteux** en termes de ressources computationnelles.
L'automatisation de l'apprentissage des représentations a aussi réduit le besoin d'expertise humaine sur l’ingénierie des données, facilitant l’adoption à grande échelle.


## Exercice 2 : Architecture LLM et scénarios d'application



| **Architecture** | **Différences architecturales**                                                                                                                           | **Application concrète**                                            | **Explication de l'adaptation**                                                                                                                                            |
| ---------------- | --------------------------------------------------------------------------------------------------------------------------------------------------------- | ------------------------------------------------------------------- | -------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **BERT**         | **Bidirectionnel**, utilise la **modélisation du langage masqué (MLM)** : le modèle voit à gauche et à droite d’un mot pour le prédire.                   | **Classification de texte** (ex : analyse de sentiments)            | La bidirectionnalité permet de capturer un contexte complet autour du texte, idéal pour comprendre le sens global et effectuer des classifications précises.               |
| **GPT**          | **Unidirectionnel (ou causale)**, **modélisation du langage causal** : prédit le prochain mot en se basant uniquement sur le contexte précédent.          | **Génération de texte libre** (ex : rédaction d’articles, chatbots) | L'architecture causale est optimisée pour générer du texte cohérent, mot après mot, ce qui est crucial pour des tâches de génération continue et naturelle.                |
| **T5**           | **Bidirectionnel** avec une approche **"text-to-text"** : toute tâche (classification, traduction, résumé) est convertie en tâche de génération de texte. | **Résumé automatique**                                              | Le format text-to-text permet à T5 de formuler des réponses flexibles et de condenser efficacement les informations tout en tenant compte du contexte complet du document. |

**Résumé** :

* **BERT** = compréhension précise (classifications).
* **GPT** = génération fluide (création de texte).
* **T5** = flexibilité multi-tâches (résumé, traduction).


## Exercice 3 : Les avantages et les considérations éthiques de la préformation



### 1. **Cinq avantages des modèles pré-entraînés** :

* **Généralisation améliorée** : les modèles pré-entraînés capturent des structures linguistiques générales, ce qui permet de mieux traiter des tâches variées, même avec peu d'exemples spécifiques.
* **Besoin réduit de données étiquetées** : après préformation, peu de données annotées suffisent pour obtenir de bonnes performances, réduisant les coûts d'annotation.
* **Réglage plus rapide** : le fine-tuning se fait rapidement sur des tâches spécifiques, car le modèle a déjà appris des connaissances générales.
* **Apprentissage par transfert** : un modèle pré-entraîné sur une grande quantité de données peut être adapté facilement à plusieurs tâches ou langues.
* **Robustesse** : grâce à une exposition massive à des données diversifiées, les modèles sont généralement plus stables face aux variations de texte (synonymes, paraphrases).

---

### 2. **Préoccupations éthiques potentielles** :

* **Biais** : les modèles peuvent intégrer et amplifier des stéréotypes présents dans les données d'entraînement.
* **Désinformation** : les LLM peuvent produire des informations incorrectes ou inventées avec un ton convaincant.
* **Utilisation abusive** : les modèles peuvent être utilisés pour des campagnes de désinformation, du spam, ou des arnaques automatisées.

---

### 3. **Stratégies d’atténuation** :

* **Filtrage et nettoyage des données** avant préformation pour limiter les biais et contenus toxiques.
* **Audit régulier des modèles** avec des jeux de données d’évaluation éthique (biais, toxicité).
* **Mécanismes de transparence** : rendre public le type de données utilisées et les limites du modèle.
* **Ajout de garde-fous techniques** : modération automatisée, filtres d’usage, limites d’accès pour éviter les abus.
* **Formation continue des utilisateurs et développeurs** pour un usage responsable.


## Exercice 4 : Plongée en profondeur dans l'architecture des transformateurs

### 1. Auto-attention et attention multi-têtes

#### a) **Auto-attention**

L’**auto-attention** permet à chaque mot d’une phrase de pondérer l’importance des autres mots pour produire une représentation enrichie.
Fonctionnement :

* Chaque mot est transformé en **query (Q)**, **key (K)** et **value (V)** via des matrices de poids.
* On calcule un score de similarité entre la **query** d’un mot et les **keys** de tous les autres mots.
* Ces scores sont normalisés (softmax), puis utilisés pour pondérer les **values**.
* Résultat : chaque mot est représenté par une combinaison pondérée de tous les mots de la séquence.

#### b) **Attention multi-têtes**

Au lieu d’utiliser une seule tête d’attention, le modèle en utilise plusieurs (multi-head), chacune ayant ses propres Q, K, V.

* **Avantages** :

  * Permet de capturer plusieurs types de relations simultanément (syntaxiques, sémantiques, longues distances).
  * Apprentissage plus riche et plus stable.

### c) **Exemple concret** :

Phrase : « Le chat qui dort est noir. »

* **Tête 1** se concentre sur la relation **"chat" ↔︎ "dort"** (sujet-verbe).
* **Tête 2** capte **"chat" ↔︎ "noir"** (relation sujet-adjectif).
* **Tête 3** se concentre sur **"qui" ↔︎ "dort"** (relation de proposition relative).

---

### 2. Objectifs de pré-formation

#### a) **MLM vs. CLM**

| Aspect       | MLM (Masked Language Modeling)          | CLM (Causal Language Modeling)            |
| ------------ | --------------------------------------- | ----------------------------------------- |
| **Principe** | Masquer des mots et prédire ces masques | Prédire le mot suivant de manière causale |
| **Contexte** | Contexte complet gauche et droite       | Contexte uniquement gauche (passé)        |
| **Exemples** | BERT, RoBERTa                           | GPT, GPT-2, GPT-3                         |

#### b) **Scénarios** :

* **MLM** : Idéal pour **classification ou extraction d’information** car le modèle comprend tout le contexte (ex : détection d’informations médicales dans un texte).
* **CLM** : Idéal pour **génération de texte** (ex : rédaction automatique d’articles).

#### c) **NSP dans BERT** :

* **NSP (Next Sentence Prediction)** aidait à apprendre les relations inter-phrases.
* **Problème** : NSP ajoutait du bruit inutile et n'améliorait pas significativement les performances.
* **Modèles modernes** comme RoBERTa l’ont supprimé au profit d’un simple MLM sur des séquences plus longues, ce qui améliore la convergence et la performance.

---

### 3. Sélection du modèle Transformer

| **Application**                   | **Modèle conseillé**       | **Justification**                                                                                  |
| --------------------------------- | -------------------------- | -------------------------------------------------------------------------------------------------- |
| **Analyse de sentiments**         | **Encodeur seul (BERT)**   | Tâche de classification, contexte complet nécessaire pour décision précise.                        |
| **Chatbot créatif**               | **Décodeur seul (GPT)**    | Tâche de génération, important de générer du texte fluide en se basant sur le passé uniquement.    |
| **Traduction anglais → espagnol** | **Encodeur-Décodeur (T5)** | Traduction = encoder la phrase source complètement, puis générer la phrase cible séquentiellement. |

---

### 4. Codage positionnel

#### a) **But** :

Les transformateurs ne tiennent pas compte de l’ordre des mots par défaut.
Le **codage positionnel** injecte des informations sur la **position** des mots dans la séquence via des vecteurs ajoutés aux embeddings.

#### b) **Importance** :

Sans codage positionnel, le modèle traiterait la phrase « Le chat mange » de la même façon que « Mange chat le » → **perte totale de la structure syntaxique**.

#### c) **Exemple concret** :

Phrase : « La voiture rouge dépasse la moto noire. »
Sans codage positionnel, le modèle pourrait ne pas comprendre qui dépasse qui, car l’ordre **sujet-verbe-objet** est ignoré.

## Exercice 6 : Température Softmax - Le régulateur d'aléatoire

---

### 1. **Scénarios de température**

| **Température** | **Comportement de sortie**                                                                                                                           |
| --------------- | ---------------------------------------------------------------------------------------------------------------------------------------------------- |
| **0,2**         | Sorties très **déterministes** : le modèle choisit presque toujours les mots avec la plus forte probabilité → discours très prévisible, peu créatif. |
| **1,5**         | Sorties très **aléatoires** : plus de diversité, mais risques d'incohérences ou d'erreurs logiques, phrases moins contrôlées.                        |
| **1**           | **Équilibre** : compromis entre créativité et cohérence, comportement standard du modèle sans sur-ajustement.                                        |

---

### 2. **Conception d'applications**

### a) **Histoires pour enfants**

* **Température recommandée : 0,8 à 1,2**
* Objectif : générer des histoires **créatives mais compréhensibles**.
* Approche :

  * **Température élevée (\~1,2)** pour stimuler la **créativité** et éviter des histoires trop répétitives.
  * **Modulation dynamique** : température plus haute pour les parties imaginatives, plus basse (0,7-0,8) pour les conclusions ou résumés cohérents.

#### b) **Résumé financier**

* **Température recommandée : 0,3 à 0,5**
* Objectif : produire des résumés **précis, cohérents et factuels**.
* Approche :

  * Baisser la température garantit que le modèle reste proche des informations les plus probables et évite les hallucinations textuelles.

---

### 3. **Température et biais**

#### a) **Effet sur les biais** :

* **Température basse** : renforce les biais présents dans le modèle, car seuls les mots les plus "probables" (selon l'entraînement) sont choisis → risque d’amplification des stéréotypes.
* **Température haute** : peut **atténuer certains biais en introduisant plus de diversité**, mais augmente aussi le risque de générer des incohérences ou du contenu problématique.

#### b) **Exemple concret** :

* **Chatbot professionnel avec température 0,2** : risque élevé de réponses stéréotypées sur certains groupes sociaux.
* **Même chatbot avec température 1,2** : propos plus variés, mais parfois incohérents ou inappropriés.

**Conclusion** : la température doit être **finement ajustée selon la tâche** pour gérer à la fois **biais et cohérence**.
