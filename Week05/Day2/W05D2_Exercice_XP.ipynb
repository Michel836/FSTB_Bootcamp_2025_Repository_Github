### Objectif:

###  **Réseau Neuronal Artificiel (RNA) - Schéma ASCII**

```
    Couche d'entrée            Couche cachée             Couche de sortie
   (features d'entrée)        (calculs internes)         (prédictions finales)

       x₁   x₂   x₃                 h₁   h₂   h₃   h₄               y₁   y₂
        ⬤    ⬤    ⬤                 ⬤    ⬤    ⬤    ⬤                 ⬤    ⬤
         \  |  /                   /|\  /|\  /|\  /|\                / \
          \ | /                  / | \/ | \/ | \/ | \              /   \
           \|/                 poids + biais + activation        poids + activation
          [W,b]             →     ReLU / Sigmoid / etc.       →     Softmax / Sigmoid
```

---

###  **Composants étiquetés :**

* **⬤** : Neurones
* **\[W,b]** : Poids (W) et biais (b)
* **→** : Flux de l'information
* **Fonctions d’activation** : appliquées entre les couches (ReLU, Sigmoid, Softmax…)

---

###  **Description du flux d'information (3–5 phrases)**

L'information entre dans le réseau par la **couche d'entrée**, composée ici de 3 neurones, chacun représentant une caractéristique (feature) d'entrée. Chaque entrée est **pondérée par un poids** et **décalée par un biais**, puis transmise à chaque neurone de la **couche cachée**. Ces neurones appliquent une **fonction d'activation** (ex. ReLU) pour introduire de la non-linéarité. Les sorties intermédiaires sont ensuite combinées et envoyées à la **couche de sortie**, où une autre activation (souvent Softmax ou Sigmoid) produit la **prédiction finale**. Ce flux est **unidirectionnel** (feedforward), des entrées vers la sortie.



## Exercice 3 : Création de l’ensemble de données et visualisation des données


### Objectif : générer un ensemble de données avec du bruit et le visualiser.
Instructions:
1. Importez les bibliothèques nécessaires : numpy, matplotlib.pyplot, et mean_squared_errorà partir de sklearn.metrics.

2. Créez un ensemble de 20 points en utilisant la formule ( y = -x^2 ), en ajoutant du bruit normalement distribué avec une moyenne de 0 et un écart type de 0,05 en utilisant np.random.normal.

Indice:

np.random.seed(0)
x = np.arange(-1, 1, 0.1)
y = -x**2 + np.random.normal(0, 0.05, len(x))


3. Tracez les points générés à l'aide d'un nuage de points pour visualiser l'ensemble de données avec du bruit.

4. Divisez l'ensemble de données en un ensemble d'entraînement (12 premiers points) et un ensemble de test (8 derniers points).

````python
# 📦 1. Importation des bibliothèques
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
````

````python
# 🔧 2. Génération des données avec bruit
np.random.seed(0)
x = np.arange(-1, 1, 0.1)  # 20 points entre -1 et 0.9
y = -x**2 + np.random.normal(0, 0.05, len(x))  # fonction cible + bruit
````

````python
# ✂️ 3. Séparation en train / test
x_train, x_test = x[:12], x[12:]
y_train, y_test = y[:12], y[12:]
````

````python
# 📊 4. Visualisation initiale des données bruitées
plt.figure(figsize=(6, 4))
plt.scatter(x, y, color='blue', label='Données bruitées')
plt.title("Nuage de points : y = -x² + bruit")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()
````

Ce **nuage de points** montre la fonction **quadratique inversée** `y = -x²` avec un **bruit aléatoire** ajouté :

---

###  **Observations :**

* La **forme générale** est bien celle d’une **parabole inversée** (courbée vers le bas).
* Le **sommet est proche de x = 0**, cohérent avec `-x²`.
* Le **bruit est modéré** (σ = 0.05) : les points ne s’écartent pas excessivement.
* Cela constitue un **bon jeu de données** pour tester des modèles simples et complexes.

---

###  **Utilité :**

Ce type de visualisation permet :

* De **valider visuellement** que les données suivent une structure attendue.
* D’estimer **s’il faut un modèle linéaire ou non** (ici : non linéaire).
* D’identifier rapidement **le niveau de bruit** présent.

---



````python
# 🔁 5. Fonction pour entraîner et prédire avec régression polynomiale
def fit_poly_model(x_train, y_train, x_test, degree):
    poly = PolynomialFeatures(degree)
    x_train_poly = poly.fit_transform(x_train.reshape(-1, 1))
    x_test_poly = poly.transform(x_test.reshape(-1, 1))
    model = LinearRegression().fit(x_train_poly, y_train)
    y_train_pred = model.predict(x_train_poly)
    y_test_pred = model.predict(x_test_poly)
    return model, poly, y_train_pred, y_test_pred
````

````python
# 📐 6. Apprentissage pour degrés 1, 2 et 10
degrees = [1, 2, 10]
models = {}
errors_train, errors_test = [], []

for deg in degrees:
    model, poly, y_train_pred, y_test_pred = fit_poly_model(x_train, y_train, x_test, deg)
    y_pred_range = model.predict(poly.transform(np.linspace(-1, 1, 100).reshape(-1, 1)))
    models[deg] = (model, poly, y_pred_range)
    errors_train.append(mean_squared_error(y_train, y_train_pred))
    errors_test.append(mean_squared_error(y_test, y_test_pred))
````

````python
# 🖼️ 7. Visualisation des courbes de régression pour chaque degré
x_range = np.linspace(-1, 1, 100)

plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='black', label='Données')
for deg in degrees:
    _, _, y_pred_range = models[deg]
    style = '--' if deg == 1 else ':' if deg == 10 else '-'
    plt.plot(x_range, y_pred_range, label=f"Degré {deg}", linestyle=style)
plt.title("Modèles polynomiaux : comparaison")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
````

Cette figure montre l'ajustement de **trois modèles de régression polynomiale** (degrés 1, 2, 10) sur les données bruitées.

---

###  **Observations :**

* **Degré 1 (ligne bleue pointillée)** :
  → **Sous-ajustement évident**
  → Ligne droite, incapable de suivre la courbure des données.

* **Degré 2 (ligne orange)** :
  → **Bon ajustement**
  → Suit la parabole avec un bon compromis entre biais et variance.

* **Degré 10 (ligne verte pointillée)** :
  → **Surajustement extrême**
  → La courbe explose vers +∞ à droite (effet typique d’un polynôme de haut degré mal régularisé).

---

###  **Conclusion visuelle immédiate :**

*  Le modèle **de degré 2** est le **meilleur choix** ici.
*  Le **degré 10** est **instable** et **sensible au bruit**, typique du **surapprentissage**.



````python
# 📉 8. Visualisation des erreurs d'entraînement et de test (MSE)
x_pos = np.arange(len(degrees))
width = 0.35

plt.figure(figsize=(6, 4))
plt.bar(x_pos - width/2, errors_train, width, label='Train', color='skyblue')
plt.bar(x_pos + width/2, errors_test, width, label='Test', color='salmon')
plt.xticks(x_pos, [f'Degré {d}' for d in degrees])
plt.ylabel("Erreur quadratique moyenne (MSE)")
plt.title("Erreur d'entraînement vs test")
plt.legend()
plt.grid(axis='y')
plt.show()
````

Ce graphique montre clairement la **performance numérique** des modèles polynomiaux.

---

### **Lecture par degré :**

* **Degré 1** :
  🔹 Erreur modérée en train & test → modèle **trop simple**, sous-ajuste.
  🔸 Confirme le **sous-ajustement** déjà vu visuellement.

* **Degré 2** :
   Erreur faible en train **et** en test → **modèle optimal**.
   Équilibre parfait entre complexité et généralisation.

* **Degré 10** :
   MSE test **catastrophique** (> 600 millions)
   Exemple clair de **surajustement** :

  * Erreur presque nulle en train,
  * Explosion de l'erreur en test.

---

###  **Conclusion globale :**

* **Modèle de degré 2** est **le meilleur compromis** pour ces données.
* Le degré 10 est un avertissement clair contre le **surapprentissage**.
* Ce graphique **confirme mathématiquement ce que les courbes de régression montrent visuellement.**


## Exercice 4 : Ajustement de modèles polynomiaux de degrés différents


Objectif : Ajuster des modèles polynomiaux de degrés variables et observer le surajustement.
Instructions:

3. Utilisez plot_polyfitpour visualiser l'ajustement pour les polynômes de degré 1, 7 et 11.

4. Observez comment le modèle se comporte sur les ensembles d'apprentissage et de test à mesure que le degré du polynôme augmente.

````python
# Imports nécessaires
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error

# 1. Fonction d'ajustement d’un polynôme de degré donné
# Elle transforme les données selon le degré, puis ajuste un modèle de régression linéaire
def polynomial_fit(x_train, y_train, degree):
    poly = PolynomialFeatures(degree)  # transforme x → [1, x, x², ..., x^degree]
    x_poly = poly.fit_transform(x_train.reshape(-1, 1))  # reshape pour sklearn
    model = LinearRegression()
    model.fit(x_poly, y_train)  # apprentissage
    return model, poly  # on retourne aussi la transformation pour les prédictions

# 2. Fonction de tracé du modèle avec les points d'entraînement et de test
# Affiche la courbe ajustée et les erreurs MSE
def plot_polyfit(x_train, y_train, x_test, y_test, degree):
    model, poly = polynomial_fit(x_train, y_train, degree)

    # Génère une plage continue pour lisser la courbe du polynôme
    x_range = np.linspace(-1, 1, 300)
    x_range_poly = poly.transform(x_range.reshape(-1, 1))
    y_pred = model.predict(x_range_poly)  # prédiction lissée pour affichage

    # Prédictions pour évaluer l'erreur sur les vraies données
    y_train_pred = model.predict(poly.transform(x_train.reshape(-1, 1)))
    y_test_pred = model.predict(poly.transform(x_test.reshape(-1, 1)))

    # Calcul des erreurs quadratiques moyennes
    train_mse = mean_squared_error(y_train, y_train_pred)
    test_mse = mean_squared_error(y_test, y_test_pred)

    # Tracé
    plt.figure(figsize=(6, 4))
    plt.scatter(x_train, y_train, color='blue', label='Entraînement')  # points d'apprentissage
    plt.scatter(x_test, y_test, color='red', label='Test')             # points de test
    plt.plot(x_range, y_pred, color='green', label=f'Degré {degree}')  # courbe ajustée
    plt.title(f"Degré {degree} - MSE train: {train_mse:.4f}, test: {test_mse:.4f}")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.legend()
    plt.grid(True)
    plt.show()

# 3. Génération des données avec bruit
# Fonction cible : y = -x² + bruit normal
np.random.seed(0)
x = np.arange(-1, 1, 0.1)  # 20 points de -1 à 0.9
y = -x**2 + np.random.normal(0, 0.05, len(x))  # bruit N(0, 0.05)

# Séparation en entraînement (12 premiers) et test (8 derniers)
x_train, x_test = x[:12], x[12:]
y_train, y_test = y[:12], y[12:]

# 4. Visualisation pour 3 degrés de complexité croissante
# Degré 1 : sous-ajustement, Degré 7 : bon compromis, Degré 11 : surajustement
for deg in [1, 7, 11]:
    plot_polyfit(x_train, y_train, x_test, y_test, deg)
````

### Degré 1 (MSE train : 0.0100, test : 1.0487)

* **Comportement** : ligne droite.
* **Résultat** : mauvais ajustement aux données courbes.
* **Symptôme** : MSE test élevé.
* **Conclusion** : **sous-ajustement** clair, modèle trop simple.

---

### Degré 7 (MSE train : 0.0008, test : 14.5851)

* **Comportement** : la courbe épouse fortement les points d'entraînement, mais **déraille** totalement en test.
* **Symptôme** : MSE test très élevé malgré un MSE train très bas.
* **Conclusion** : **surajustement**, le modèle s’adapte trop aux données de train.

---

### Degré 11 (MSE train : 0.0000, test : > 1e12)

* **Comportement** : instabilité extrême, explosion des valeurs.
* **Résultat** : prédictions incohérentes et inutilisables.
* **Conclusion** : **surajustement extrême**, avec comportement numérique instable (effet Runge).

---

### Conclusion globale

* **Modèle de degré 1** : trop simple, ne capture pas la forme de la fonction.
* **Modèles de degré 7 et 11** : trop complexes, capturent le bruit et échouent à généraliser.
* **Modèle optimal** (vu précédemment) : autour de **degré 2 à 3**.


## 🌟 Exercice 5 : Validation croisée pour trouver le degré optimal


Objectif : Utiliser la validation croisée pour déterminer le meilleur degré polynomial.
Instructions:
1. Parcourez les degrés polynomiaux de 1 à 11.
2. Pour chaque degré, calculez l'erreur quadratique moyenne (RMSE) pour les ensembles d'entraînement et de test.
3. Stockez les résultats (degré, RMSE pour l'entraînement, RMSE pour le test) dans une liste.
4. Tracez les valeurs RMSE pour les ensembles d'entraînement et de test par rapport au degré polynomial en utilisant une échelle logarithmique pour l'axe des y.
5. Identifiez le degré qui minimise le RMSE sur l'ensemble de test et confirmez qu'il correspond au véritable modèle sous-jacent (( y = -x^2 )).

````python
# Imports nécessaires
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error

# Données (y = -x² + bruit)
np.random.seed(0)
x = np.arange(-1, 1, 0.1)
y = -x**2 + np.random.normal(0, 0.05, len(x))
x_train, x_test = x[:12], x[12:]
y_train, y_test = y[:12], y[12:]

# Stockage des résultats
results = []

# Boucle sur les degrés de 1 à 11
for degree in range(1, 12):
    # Transformation polynomiale
    poly = PolynomialFeatures(degree)
    x_train_poly = poly.fit_transform(x_train.reshape(-1, 1))
    x_test_poly = poly.transform(x_test.reshape(-1, 1))
    
    # Régression
    model = LinearRegression().fit(x_train_poly, y_train)
    
    # Prédictions
    y_train_pred = model.predict(x_train_poly)
    y_test_pred = model.predict(x_test_poly)
    
    # Calcul RMSE (racine MSE)
    rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
    rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))
    
    # Stockage
    results.append((degree, rmse_train, rmse_test))

# Extraction des valeurs pour tracé
degrees = [r[0] for r in results]
rmse_train_vals = [r[1] for r in results]
rmse_test_vals = [r[2] for r in results]

# Tracé des courbes RMSE en fonction du degré
plt.figure(figsize=(7, 5))
plt.plot(degrees, rmse_train_vals, label='RMSE entraînement', marker='o')
plt.plot(degrees, rmse_test_vals, label='RMSE test', marker='s')
plt.yscale('log')  # échelle logarithmique
plt.xlabel("Degré du polynôme")
plt.ylabel("RMSE (log)")
plt.title("Erreur RMSE par degré polynomial")
plt.legend()
plt.grid(True)
plt.show()

# Identification du degré optimal (minimisant RMSE test)
min_rmse_test = min(rmse_test_vals)
optimal_degree = degrees[rmse_test_vals.index(min_rmse_test)]

print(f"Degré optimal = {optimal_degree} (RMSE test minimal = {min_rmse_test:.4f})")
````

Ce graphique montre clairement l'évolution de l'erreur **RMSE** sur les ensembles **d'entraînement** et **de test** en fonction du **degré du polynôme**, en **échelle logarithmique**.

---

### Observations directes :

* **RMSE entraînement** (courbe bleue) :

  * Diminue régulièrement avec le degré.
  * Montre que le modèle devient plus souple et colle davantage aux données de train.
  * Au-delà du degré 10, la courbe devient numériquement instable (proche de zéro voire erreurs de machine).

* **RMSE test** (courbe orange) :

  * Atteint un **minimum clair autour du degré 2**, ce qui confirme que c’est **le degré optimal**.
  * Croît fortement à partir du degré 3, indiquant un **surajustement croissant**.
  * Explosion brutale pour les degrés élevés → surapprentissage extrême et instabilités.

---

### Conclusion :

* Le **degré optimal est 2**, ce qui correspond exactement à la **vraie fonction `y = -x²`**.
* Ce résultat **valide expérimentalement** le bon comportement du modèle et l’utilité de la validation croisée.
* Cette démarche permet de **choisir un modèle sans surajuster ni sous-ajuster**, même en présence de bruit.
