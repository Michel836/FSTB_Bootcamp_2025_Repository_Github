```python
# === 1. INSTALLATION (Exécuter séparément si besoin) ===
# %pip install -qU langchain==0.3.27 langchain-community==0.3.27 langgraph==0.6.2 tavily-python langchain-mistralai
```

```python
# === 2. IMPORTS ET VÉRIFICATION DES VERSIONS ===
import os, getpass
from importlib.metadata import version
import langchain
print("2️⃣ LangChain v:", langchain.__version__)           # doit être 0.3.27
print("2️⃣ LangGraph v:", version("langgraph"))          # doit être 0.6.2
```

```python
# === 3. CLÉS API (sécurisées) ===
os.environ.setdefault("LANGSMITH_TRACING", "true")
if not os.getenv("LANGSMITH_API_KEY"):
    os.environ["LANGSMITH_API_KEY"] = getpass.getpass("LangSmith API Key: ")
if not os.getenv("TAVILY_API_KEY"):
    os.environ["TAVILY_API_KEY"] = getpass.getpass("Tavily API Key: ")
if not os.getenv("MISTRAL_API_KEY"):
    os.environ["MISTRAL_API_KEY"] = getpass.getpass("Mistral API Key: ")
assert os.getenv("LANGSMITH_API_KEY") and os.getenv("TAVILY_API_KEY") and os.getenv("MISTRAL_API_KEY")
print("3️⃣ Clés API configurées")
```

```python
# === 4. INITIALISATION DU LLM Mistral (ChatMistralAI) ===
from langchain_mistralai.chat_models import ChatMistralAI
from langchain_core.messages import HumanMessage

model = ChatMistralAI(model="mistral-large-latest", api_key=os.getenv("MISTRAL_API_KEY"), temperature=0.0, max_retries=2)
resp = model.invoke([HumanMessage(content="Donne-moi un fait amusant sur l’espace.")])
print("4️⃣ Réponse LLM test:", resp.content)  # vérifie que le modèle fonctionne
```

```python
# === 5. DÉFINITION DE L’OUTIL DE RECHERCHE (Tavily) ===
from langchain_community.tools.tavily_search import TavilySearchResults

search = TavilySearchResults()
tools = [search]
res = search.run("latest AI breakthroughs in healthcare")
print("5️⃣ Extraits recherche (Tavily):", res[:2])   # vérifie l’outil fonctionne
```

```python
# === 6. BINDING DES OUTILS AU LLM ===
model_with_tools = model.bind_tools(tools)
r1 = model_with_tools.invoke([HumanMessage(content="Quelle est la capitale de la France ?")])
print("6️⃣ Sans outil – output:", r1.content, "| tool_calls:", r1.tool_calls)
r2 = model_with_tools.invoke([HumanMessage(content="Find the latest breakthroughs in AI for emergency medicine.")])
print("6️⃣ tool_calls générés:", r2.tool_calls)     # montre que l’appel outil est bien généré
```

```python
# === 7. CRÉATION DE L’AGENT ReAct (LangGraph) ===
from langgraph.prebuilt import create_react_agent

agent_executor = create_react_agent(
    model=model,
    tools=tools,
    prompt="You are a helpful assistant that can think, act with tools, and answer user questions."
)
print("7️⃣ Agent ReAct prêt:", agent_executor)
```

```python
# === 8. EXÉCUTION DE L’AGENT – ÉTATS FINAUX ===
state1 = agent_executor.invoke({"messages":[{"role":"user","content":"What is the capital of France?"}]})
print("8️⃣ State1:", state1)

state2 = agent_executor.invoke({"messages":[{"role":"user","content":"Tell me a joke about robots."}]})
print("8️⃣ State2:", state2)

state3 = agent_executor.invoke({"messages":[{"role":"user","content":"Find the latest breakthroughs in AI for emergency medicine."}]})
print("8️⃣ State3:", state3)   # inclut tool_calls et réponses
```

```python
# === 9. RÉPONSES EN STREAMING ===
msgs = [{"role":"user","content":"Find the top 3 recent AI papers on medical imaging."}]
stream_it = agent_executor.stream({"messages": msgs}, stream_mode="auto")
print("9️⃣ Streaming en direct:")
for step in stream_it:
    step["messages"][-1]["content"].pretty_print()
```

```python
# === 10. RÉCAPITULATIF DES LIVRABLES ===
print("🔟 ✅ Livrables générés :")
print("   • Agent ReAct opérationnel via LangGraph")
print("   • Recherche web via Tavily intégrée")
print("   • Traçage LangSmith actif (grâce à LANGSMITH_TRACING)")
print("   • Exécutions classiques et mode streaming")
print("   • États inspectables (inputs, outputs, tool_calls)")
```

##  Points d’attention

* **Créer l’agent avec `prompt` explicite** est requis par `create_react_agent`, faute de quoi l’appel lève une erreur (`missing model or prompt`) ([GitHub][8], [langchain-ai.github.io][6], [LangChain][1], [LangChain][5], [langchain-ai.github.io][2], [Medium][9], [Medium][10], [LangChain Forum][7]).
* **TavilySearchResults** est inclus dans `langchain_community.tools`; il offre une recherche optimisée pour les agents LLM, avec paramètres comme `max_results`, `time_range` etc. ([LangChain][3]).
* **Streaming via `.stream()`** permet de capturer les étapes intermédiaires (pensées, tool calls) là où `invoke()` ne renvoie que l’état final ([LangChain Forum][7]).
* **Traçage LangSmith**, bien qu’implémenté via variable d’environnement, nécessite un compte LangSmith actif pour accéder aux dashboards.

## Applications innovantes et creatives

### 1.  *Visites culturelles immersives gamifiées avec IA générative + AR*

Implémente une **application mobile ou web** alliant **réalité augmentée** et **IA générative narrative** pour revisiter des sites historiques (châteaux, musées, monuments). L’IA adapte en temps réel le contenu visuel et l’histoire racontée en fonction du lieu, du profil de l’utilisateur et de son parcours.

* Architecture modulaire inspirée de l’article sur les applications de patrimoine gamifiées et adaptatives ([arXiv][1]).
* Combine AR, génération d’images/mots, gamification et storytelling personnalisé.

---

### 2.  *Jeu narratif coopératif IA inspiré de « I Like Your Story! »*

Un projet de **jeu interactif co-créatif** où un agent IA (à personnalité distincte) collabore ou résiste aux utilisateurs en fonction de choix scénarisés. Le joueur raconte, l’IA infuse des contraintes, génère des rebondissements et révèle des “pouvoirs narratifs” selon la stratégie adoptée.

* Inspiré de l’expérience décrite dans l’article sur le storytelling persona-driven ([arXiv][2]).
* Favorise l’engagement créatif, le feedback ludique et une montée en puissance progressive de l’histoire.

---

### 3.  *Création de motifs textiles traditionnels enrichis par IA générative*

Développe un outil pour générer des **motifs textiles** traditionnels revisités (ex. tissage japonais) : l’utilisateur fournit une image, l’IA propose des compositions originales respectant l’esprit culturel.

* Application inspirée de la collaboration IA‑artisanat mentionnée au Japon pour préserver des motifs Nishijinori ([AP News][3]).
* Approche hybride : l’IA suggère des variantes, l’artisan valide ou ajuste manuellement.

---

### 4.  *Assistant créatif musical personnalisé pour artistes et amateurs*

Crée une plateforme où un générateur IA propose des mélodies ou pistes selon des styles ou émotions (calme, tribale, orchestral), à partir de textes ou mood boards. L’utilisateur peut interagir (éditer, changer tempo, instrument).

* Basée sur des systèmes existants comme AIVA, Suno ou MuseNet ([Wikipédia][4]).
* Enfin, intègre un module ReAct pour que l’agent puisse suggérer des modifications, transitions ou progressions harmoniques.

---

### 5.  *Assistant de soutien émotionnel et créatif (IA compagnon artistique)*

Un agent IA empathique proposant à la fois du soutien émotionnel et des suggestions artistiques (écriture, dessin, collage numérique) selon l’humeur exprimée : dialogue, conseils créatifs, prompts visuels, réflexions poétiques.

* En phase avec le besoin croissant d’**AI pour thérapie, introspection ou compagnonnage** ([learn.filtered.com][5], [Wikipédia][6], [hbr.org][7]).
* Combine agents conversationnels, génération d’images & prompts créatifs, ou de mini-exercices d’art-thérapie.
