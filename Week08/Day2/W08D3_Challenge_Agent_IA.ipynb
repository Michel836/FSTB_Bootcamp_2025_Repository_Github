## 1.  **Thérapie artistique générative personnalisée**

Fusionne un modèle texte-à-image (comme DALL·E 3 ou Stable Diffusion) avec des prompts générés par l’utilisateur pour créer des œuvres visuelles adaptées à un état émotionnel. Les utilisateurs décrivent leur humeur ou leurs émotions, l’IA génère une image expressive, puis propose des réflexions ou exercices basés sur l’œuvre générée. Inspiré de travaux comme l’intégration de l’IA dans l’art thérapie en 2024‑2025 ([McKinsey & Company][1]).

**Pédagogiquement**, ce projet nécessite traitement texte et image, fine-tuning de diffusion model, interface interactive, et interfaces utilisateur émotionnellement adaptatives.

---

## 2.  **Jeu narratif co-créatif avec agent personnage**

Une **expérience de storytelling** où un agent IA incarne un personnage avec une personnalité définie (par exemple un roi excentrique), interagit avec les utilisateurs pour élaborer une histoire commune. Ce modèle est basé sur "I Like Your Story!" (2025), où l’agent évalue les contributions du joueur, fournit des retours, et modifie la narration de manière réactive ([arXiv][2]).

**Apprentissage** : génération de texte adaptative, gestion d’état narratif, design d’agent conversationnel avec retour utilisateur.

---

## 3.  **Assistant de production vidéo créative (Agentic AI)**

Utilise des plateformes comme **Runway AI** ou **Firefly** pour aider les créateurs à générer des storyboards, transformer des images fixes en séquences animées, réaliser du compositing vidéo, retouche, etc. L’IA devient un assistant créatif, accélérant phases de pré- et post-production ([TechRadar][3], [TechRadar][4]).

**Compétences développées** : traitement multimodal image/vidéo, intégration API, interfaces UX pour workflow créatif.

---

## 4.  **Outil de prototypage de produits via design génératif**

Combine l’IA générative de design (paramétrique ou diffusion) et des contraintes prédéfinies (coût, matériaux, durabilité). Génère automatiquement des prototypes visuels ou objets 3D optimisés. Le champ correspond au **generative design** appliqué à l’architecture ou produit ([Wikipédia][5]).

**Aspects techniques** : génération multimodale, optimisation par contraintes, rendu en 3D ou images, interface utilisateur modulable.

---

## 5.  **Compagnon mental virtuel avec conversations empathiques**

Crée un chatbot basé sur des LLMs (Claude, ChatGPT, Gemini) capable de détecter l’état émotionnel de l’utilisateur (analyse NLP + tonalité) et proposer des techniques de bien-être : méditation, exercices cognitifs, suggestions narratives ou art visuel personnalisé en soutien émotionnel. Comparable à Wysa, Woebot ou XAIA ([Wikipédia][6]).

**Valeur pédagogique** : LLMs émotionnels, pipelines multimodaux (texte + possible image générative), UX sensibilisée à l’éthique.
