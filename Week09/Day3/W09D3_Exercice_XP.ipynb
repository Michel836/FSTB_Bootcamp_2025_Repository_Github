# 3. Charger le modèle et les objets de prétraitement

```python
import pickle
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Charger le modèle et les objets de prétraitement
with open("rf_model.pkl", "rb") as f:
    components = pickle.load(f)

# On récupère les objets selon leur clé exacte (adapter selon votre pickle !)
num_imputer = components["num_imputer"]
cat_imputer = components["cat_imputer"]
encoder = components["encoder"]
model = components["model"]
scaler = components["scaler"]  # StandardScaler entraîné

# Spécifier les colonnes attendues (à ajuster selon votre modèle)
categorical_columns = ['family', 'state', 'store_type', 'cluster']
numerical_columns = ['store_nbr', 'onpromotion', 'transactions', 'crude_oil_price', 'month', 'day', 'dayofweek']
```

# 4. Interface utilisateur Streamlit complète

(À mettre dans app.py puis exécuter avec streamlit run app.py)

```python
import streamlit as st

# Titre, image, légende
st.image("https://cdn.pixabay.com/photo/2016/11/29/09/08/chart-1869178_1280.jpg", use_column_width=True)
st.title("Application de prévision des ventes")
st.caption("Saisissez les caractéristiques pour prédire les ventes d’un magasin donné.")

# Sidebar explicative
st.sidebar.header("Description des champs")
st.sidebar.markdown("""
- **Store Number** : Numéro du magasin.
- **Product Family** : Catégorie de produits (AUTOMOTIVE, BEAUTY, etc.).
- **On Promotion** : Nombre d’articles en promotion.
- **State** : Région où se trouve le magasin.
- **Store Type** : Type de magasin (A, B, C, D, etc.).
- **Cluster** : Cluster géographique ou commercial du magasin.
- **Transactions** : Nombre de transactions en caisse sur la période.
- **Crude Oil Price** : Prix du pétrole brut.
- **Month** : Mois de l’année.
- **Day** : Jour du mois.
- **Day of Week** : Jour de la semaine.
""")

# Saisie des entrées utilisateur en 3 colonnes
input_data = {}
col1, col2, col3 = st.columns(3)

with col1:
    input_data['store_nbr'] = st.slider("Numéro du magasin", 1, 54, 1)
    input_data['family'] = st.selectbox("Famille de produit", ['AUTOMOTIVE', 'BEAUTY', 'FOODS', 'CLEANING', 'HOME', 'CLOTHING', 'STATIONERY', 'HARDWARE'])
    input_data['onpromotion'] = st.slider("Articles en promotion", 0, 50, 0)
    input_data['state'] = st.selectbox("État", ['Pichincha', 'Cotopaxi', 'Chimborazo', 'Imbabura', 'Santo Domingo de los Tsachilas', 'Bolivar', 'Pastaza', 'Tungurahua', 'Guayas', 'Santa Elena', 'Los Rios', 'Azuay', 'Loja', 'El Oro', 'Esmeraldas', 'Manabi'])
    input_data['transactions'] = st.number_input("Transactions", 0, 10000, 100)

with col2:
    input_data['store_type'] = st.selectbox("Type de magasin", ['A', 'B', 'C', 'D'])
    input_data['cluster'] = st.slider("Cluster", 1, 17, 1)
    input_data['crude_oil_price'] = st.number_input("Prix du pétrole brut", 10.0, 150.0, 50.0)

with col3:
    input_data['month'] = st.slider("Mois", 1, 12, 1)
    input_data['day'] = st.slider("Jour", 1, 31, 1)
    input_data['dayofweek'] = st.slider("Jour de la semaine", 0, 6, 0)
```

# 5. Pipeline de prédiction

```python
# Créer un bouton pour lancer la prédiction
if st.button("Prédire"):
    # 1. Convertir les entrées en DataFrame
    input_df = pd.DataFrame([input_data])

    # 2. (Exemple) Regroupement des familles si besoin
    # Ici, on peut faire des regroupements spécifiques sur 'family' si le modèle l'a appris
    # À adapter selon votre pipeline d’entraînement

    # 3. Séparation des colonnes
    input_df_cat = input_df[categorical_columns]
    input_df_num = input_df[numerical_columns]

    # 4. Imputation (remplissage des valeurs manquantes, si pertinent)
    input_df_cat_imputed = cat_imputer.transform(input_df_cat)
    input_df_num_imputed = num_imputer.transform(input_df_num)

    # 5. Encodage des variables catégorielles
    input_df_cat_encoded = pd.DataFrame(
        encoder.transform(input_df_cat_imputed).toarray(),
        columns=encoder.get_feature_names_out(categorical_columns)
    )

    # 6. Mise à l’échelle des variables numériques
    input_df_num_scaled = scaler.transform(input_df_num_imputed)
    input_df_num_scaled = pd.DataFrame(input_df_num_scaled, columns=numerical_columns)

    # 7. Fusion des features finales
    input_df_processed = pd.concat([input_df_cat_encoded, input_df_num_scaled], axis=1)

    # 8. Prédiction
    predictions = model.predict(input_df_processed)

    # 9. Affichage du résultat
    st.success(f"Prévision de ventes : {predictions[0]:,.2f} unités")
```
