# 2. Chargement du modèle et premiers tests

```python
# Importer les modules nécessaires
from transformers import pipeline, Conversation

# Charger le modèle conversationnel BlenderBot (facebook/blenderbot-400M-distill)
# Le pipeline 'conversational' prend en charge la gestion du contexte
chatbot = pipeline("conversational", model="facebook/blenderbot-400M-distill")

# Test simple en console pour vérifier que tout fonctionne
conv = Conversation("Bonjour, qui es-tu ?")
result = chatbot(conv)
print(result)
```

# 3. Fonction de gestion de la conversation pour Gradio

```python
# Historique géré sous forme de liste de tuples (message, réponse)
# Gradio transmet à chaque appel la liste de l'historique, ce qui permet la continuité du dialogue

def mini_chatbot(message, history):
    """
    Fonction de backend pour l'interface Gradio.
    - message : texte de l'utilisateur
    - history : liste de tuples (user_message, bot_response) gérée automatiquement par Gradio

    On reconstitue la conversation complète pour préserver le contexte.
    """
    # Extraction des anciens messages utilisateur et réponses générées
    past_user_inputs = [pair[0] for pair in history] if history else []
    generated_responses = [pair[1] for pair in history] if history else []

    # Création d'un objet Conversation avec l'historique complet
    conversation = Conversation(
        text=message,
        past_user_inputs=past_user_inputs,
        generated_responses=generated_responses
    )
    # Génération de la réponse via le modèle
    chatbot(conversation)
    # Retourne uniquement la dernière réponse générée
    return conversation.generated_responses[-1]
```

# 4. Création de l’interface Gradio et lancement du chatbot

```python
import gradio as gr

# Création de l'interface utilisateur en mode chat
demo_chatbot = gr.ChatInterface(
    mini_chatbot,
    title="Chatbot IA - BlenderBot",
    description="Discutez avec un chatbot IA basé sur BlenderBot (facebook/blenderbot-400M-distill). Entrez votre message ci-dessous."
)

# Lancement de l'interface (une fois exécuté, un lien s'affiche pour ouvrir le chat dans votre navigateur)
demo_chatbot.launch()
```
