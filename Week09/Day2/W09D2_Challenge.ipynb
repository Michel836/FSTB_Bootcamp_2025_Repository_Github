```python
# 1. Imports
import gradio as gr
from datetime import datetime

# 2. Fonction principale : génère le message d'accueil en fonction du nom, de la langue et de l'heure
def saluer_utilisateur(nom: str, langue: str) -> str:
    """
    Retourne un message d'accueil personnalisé selon le nom, la langue et l'heure du jour.
    - nom: string (nom de l'utilisateur)
    - langue: 'French' ou 'English'
    """
    if not nom or not nom.strip():
        return "Veuillez entrer un nom valide." if langue == "French" else "Please enter a valid name."
    
    heure = datetime.now().hour

    # Détection du moment de la journée
    if 5 <= heure < 12:
        moment = "morning"
    elif 12 <= heure < 18:
        moment = "afternoon"
    else:
        moment = "evening"
    
    # Table de traduction pour les messages d'accueil
    messages = {
        "French": {
            "morning": "Bonjour",
            "afternoon": "Bon après-midi",
            "evening": "Bonsoir"
        },
        "English": {
            "morning": "Good morning",
            "afternoon": "Good afternoon",
            "evening": "Good evening"
        }
    }

    # Sécurité sur la langue (fallback anglais)
    if langue not in messages:
        langue = "English"
    
    message_accueil = messages[langue][moment]
    # Message final personnalisé
    return f"{message_accueil}, {nom.strip()} !"

# 3. Création de l'interface Gradio
with gr.Blocks() as demo:
    gr.Markdown("## Mini Greeter App — Application d'accueil personnalisée")
    with gr.Row():
        nom = gr.Textbox(label="Votre nom", placeholder="Entrez votre nom ici…")
        langue = gr.Dropdown(label="Langue", choices=["French", "English"], value="French")
    bouton = gr.Button("Obtenir mon message d'accueil")
    sortie = gr.Textbox(label="Message d'accueil", interactive=False)
    exemples = gr.Examples(
        examples=[["Alice", "English"], ["Jean", "French"]],
        inputs=[nom, langue]
    )

    # Lien fonctionnel : au clic, exécute la fonction saluer_utilisateur
    bouton.click(fn=saluer_utilisateur, inputs=[nom, langue], outputs=sortie)

# 4. Lancement de l'application (en local)
demo.launch()
```
