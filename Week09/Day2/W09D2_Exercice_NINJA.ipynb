# Exercice 1 – Gradio + FastAPI + LangChain : Assistant IA modulaire

## Cellule 1 — Backend FastAPI

Lance ce backend dans un terminal séparé (uvicorn backend:app --reload) ou via une cellule Jupyter utilisant %run ou un shell si tu as besoin.

```python
# backend.py
# Backend FastAPI pour router les requêtes des différents onglets de l'assistant IA

from fastapi import FastAPI
from pydantic import BaseModel

# Simulations LLM, search et calcul pour démonstration
def dummy_llm_response(message):
    return f"LLM: J'ai reçu : {message}"

def dummy_doc_search(query):
    return f"Recherche documentaire simulée pour : {query}"

def dummy_math_eval(expr):
    try:
        result = eval(expr)
        return f"Résultat: {result}"
    except Exception as e:
        return f"Erreur de calcul: {e}"

app = FastAPI()

class AssistantRequest(BaseModel):
    username: str
    tab: str
    content: str

@app.post("/assistant")
def assistant(req: AssistantRequest):
    if req.tab == "chat":
        response = dummy_llm_response(req.content)
    elif req.tab == "recherche":
        response = dummy_doc_search(req.content)
    elif req.tab == "math":
        response = dummy_math_eval(req.content)
    else:
        response = "Onglet inconnu"
    return {"username": req.username, "tab": req.tab, "content": req.content, "response": response}
```

Explications :

* Trois fonctions simulées pour LLM, recherche documentaire et calcul mathématique.

* Route POST /assistant qui reçoit la requête, route sur le bon outil et retourne la réponse.

## Cellule 2 — Frontend Gradio multi-onglets

```python
# frontend_gradio.py
# Application Gradio avec trois onglets connectés au backend FastAPI
import gradio as gr
import requests

# Adresse du backend FastAPI
API_URL = "http://localhost:8000/assistant"

def send_to_backend(tab, username, content):
    payload = {"username": username, "tab": tab, "content": content}
    r = requests.post(API_URL, json=payload)
    return r.json()["response"]

with gr.Blocks() as demo:
    gr.Markdown("# Assistant IA Modulaire (XP Ninja)")
    with gr.Row():
        username = gr.Textbox(label="Nom d'utilisateur", value="Chu", interactive=True)
    with gr.Tabs():
        with gr.TabItem("Chat"):
            chat_input = gr.Textbox(label="Message au LLM")
            chat_output = gr.Textbox(label="Réponse", interactive=False)
            chat_btn = gr.Button("Envoyer")
            chat_btn.click(
                lambda u, c: send_to_backend("chat", u, c),
                inputs=[username, chat_input],
                outputs=chat_output
            )
        with gr.TabItem("Recherche"):
            search_input = gr.Textbox(label="Recherche documentaire")
            search_output = gr.Textbox(label="Résultat", interactive=False)
            search_btn = gr.Button("Rechercher")
            search_btn.click(
                lambda u, c: send_to_backend("recherche", u, c),
                inputs=[username, search_input],
                outputs=search_output
            )
        with gr.TabItem("Math"):
            math_input = gr.Textbox(label="Expression mathématique (ex : 2+2*5)")
            math_output = gr.Textbox(label="Résultat", interactive=False)
            math_btn = gr.Button("Calculer")
            math_btn.click(
                lambda u, c: send_to_backend("math", u, c),
                inputs=[username, math_input],
                outputs=math_output
            )

demo.launch()
```

Explications :

* Trois onglets : Chat (LLM), Recherche documentaire, Math.

* Le nom d’utilisateur est partagé.

* Chaque onglet envoie les données via POST au backend FastAPI et affiche la réponse.

* Tu peux étendre en ajoutant un quatrième onglet « Historique » qui affiche un état partagé.

# Exercice 2 – Application Streamlit avec secrets, mémoire, API externe
Cellule 3 — Application Streamlit simplifiée

Crée un fichier streamlit_app.py et lance-le avec streamlit run streamlit_app.py.
La version Jupyter « pure » n’est pas supportée par Streamlit, mais tu peux t’inspirer du code pour adaptation.

```python
import streamlit as st
import requests

# Simuler le stockage sécurisé des clés dans .streamlit/secrets.toml
API_KEY = st.secrets["weather_api_key"] if "weather_api_key" in st.secrets else "clé_fake"

def get_weather(city):
    # Simulation d'appel API météo (remplace par une vraie API si besoin)
    if city.lower() == "paris":
        return {"temp": 22, "condition": "Ensoleillé"}
    else:
        return {"temp": 18, "condition": "Variable"}

def summarize_weather(data):
    return f"Il fait {data['temp']}°C et le temps est {data['condition']}."

st.sidebar.title("Options")
theme = st.sidebar.radio("Thème", options=["Clair", "Sombre"])
if st.sidebar.button("Réinitialiser le chat"):
    st.session_state["history"] = []

if "history" not in st.session_state:
    st.session_state["history"] = []

st.title("Assistant IA Streamlit")

user_input = st.text_input("Posez votre question (ex : météo à Paris)")
if user_input:
    st.session_state["history"].append(("user", user_input))
    if "météo" in user_input.lower():
        city = user_input.split("à")[-1].strip() if "à" in user_input else "Paris"
        meteo = get_weather(city)
        response = summarize_weather(meteo)
    else:
        response = f"LLM : Je traite votre demande '{user_input}'..."
    st.session_state["history"].append(("assistant", response))

# Afficher l'historique
for role, msg in st.session_state["history"]:
    st.write(f"**{role.capitalize()}** : {msg}")

# Bonus : export CSV
if st.button("Exporter l'historique (CSV)"):
    import pandas as pd
    df = pd.DataFrame(st.session_state["history"], columns=["role", "message"])
    df.to_csv("chat_history.csv", index=False)
    st.success("Exporté dans chat_history.csv")
```

Explications :

* Simule une API météo (remplace par une vraie clé/API si besoin).

* Utilise st.session_state pour stocker l’historique.

* Affiche l’historique en distinguant les rôles, permet un export CSV.

# Exercice 3 – Agent LangChain + Gradio (raisonnement multi-outils)
Cellule 4 — Agent LangChain avec mémoire et sélection d’outil

Suppose que tu as installé langchain, wikipedia, faiss-cpu…
Ceci est un exemple pédagogique ; adapte à ton backend/documents.

```python
# agent_gradio.py
# Agent LangChain avec outils, mémoire et interface Gradio

import gradio as gr
from langchain.llms import OpenAI
from langchain.tools import WikipediaQueryRun, Tool
from langchain.memory import ConversationBufferMemory
from langchain.agents import initialize_agent

# Outils : calcul, recherche wikipédia
def calculatrice(expr):
    try:
        return str(eval(expr))
    except Exception:
        return "Erreur de calcul"

wiki_tool = Tool(
    name="Wikipedia",
    func=WikipediaQueryRun(api_wrapper=None),
    description="Recherche Wikipedia"
)
calc_tool = Tool(
    name="Calculatrice",
    func=calculatrice,
    description="Effectuer des calculs arithmétiques"
)

tools = [wiki_tool, calc_tool]
memory = ConversationBufferMemory(memory_key="chat_history")
llm = OpenAI(temperature=0)

agent = initialize_agent(
    tools=tools,
    llm=llm,
    agent="zero-shot-react-description",
    verbose=True,
    memory=memory
)

def agent_interact(user_input, history):
    # Ajout à la mémoire
    response = agent.run(user_input)
    # Détection de l’outil utilisé (simplifié, voir logs pour vraie extraction)
    if "wikipedia" in response.lower():
        outil = "Wikipedia"
        explication = "Utilisation de Wikipedia car entité nommée détectée."
    elif any(op in user_input for op in "+-*/"):
        outil = "Calculatrice"
        explication = "Calcul détecté dans la requête."
    else:
        outil = "LLM seul"
        explication = "Réponse générée directement."
    history = history or []
    history.append((user_input, response, outil, explication))
    return history, history

with gr.Blocks() as demo:
    gr.Markdown("# Agent IA multi-outils (XP Ninja)")
    chatbot = gr.Dataframe(headers=["Question", "Réponse", "Outil", "Explication"])
    inp = gr.Textbox(label="Posez votre question")
    state = gr.State([])
    btn = gr.Button("Envoyer")
    btn.click(agent_interact, [inp, state], [chatbot, state])

demo.launch()
```

Explications :

* Initialise un agent LangChain avec deux outils : calculatrice et recherche Wikipedia.

* La logique détecte l’outil utilisé (de façon basique ici).

* Affiche un tableau (question, réponse, outil, explication).
