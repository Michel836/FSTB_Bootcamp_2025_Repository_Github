Qu'est-ce qu'un ensemble de données de type SQuAD ?
SQuAD signifie Stanford Question Answering Dataset. Il s'agit d'un ensemble de données de référence utilisé pour entraîner et évaluer des modèles de questions-réponses extractives, un type d'assurance qualité où le modèle sélectionne comme réponse un fragment de texte (une séquence continue de mots) d'un passage donné.



💡 Que signifie « style SQuAD »
Un jeu de données de type SQuAD suit la même structure et la même logique que le jeu de données SQuAD original. Il contient généralement :

Un contexte (paragraphe) : un passage de texte où l’on peut trouver la réponse à la question.
Une question : une requête en langage naturel sur le contexte.
Réponse : une portion de texte textuelle présente textuellement dans le contexte. Le modèle apprend à la trouver et à l'extraire.
L'index de départ : la position (index de caractère) où commence la réponse dans le contexte. Cela aide les modèles à savoir précisément où pointer dans le texte.

# Analyse détaillée et une **approche de résolution simple, robuste et fonctionnelle** pour le défi “Affiner un modèle de réponse aux questions avec Haystack”:

---

## 1. **Analyse en profondeur de l’énoncé**

**But du projet**
Construire un système de questions-réponses (QA) extractif, entraîné et affiné sur un domaine spécifique, en s’appuyant sur Haystack, un framework open source dédié au NLP, et des modèles de type transformer (comme RoBERTa SQuAD) pour améliorer la pertinence des réponses à partir de documents.

**Tâches principales**

**Contraintes / Spécificités**

* Utilisation de Google Colab (avec GPU)

* Manipulation de données au format SQuAD JSON

**Points d’apprentissage**

* Préparation de l’environnement d’entraînement (GPU, installation Haystack)
* Préparation, formatage et chargement de données SQuAD
* Fine-tuning (affinage) d’un modèle QA existant
* Sauvegarde, chargement et test du modèle affiné
* Évaluation comparative (avant/après fine-tuning)

---

## 2. **Approche de résolution**

### **Étape 1 – Préparation de l’environnement**

* **Outil** : Google Colab (exécution GPU, installation des librairies via pip)
* **Action** : Installation de Haystack et de toutes les dépendances (torch, transformers…)
* **Bonne pratique** : Vérifier le GPU disponible, configurer le niveau de logs à INFO pour le debug

---

### **Étape 2 – Données SQuAD**

* **Source** : Télécharger un dataset SQuAD officiel ou un dataset spécifique au domaine, ou en créer un personnalisé
* **Format** : S’assurer que le JSON respecte bien la structure attendue (context, question, answers\[text+start])
* **Stockage** : Sauvegarder les fichiers dans un répertoire /data dans l’environnement Colab

---

### **Étape 3 – Chargement du modèle de base**

* **Choix du modèle** : Prendre un modèle extractif pré-entraîné sur SQuAD (ex : `deepset/roberta-base-squad2`)
* **Composant Haystack** : Initialiser un `FARMReader` avec ce modèle

---

### **Étape 4 – Fine-tuning (affinage) du modèle**

* **Méthode** : Utiliser l’API de Haystack pour entraîner le modèle sur les nouvelles données
* **Paramètres** : Définir les epochs, batch size, learning rate adaptés aux ressources GPU/temps
* **Monitoring** : Suivre la progression via la journalisation, ajuster si surapprentissage ou sous-apprentissage détecté

---

### **Étape 5 – Sauvegarde et rechargement du modèle**

* **Action** : Enregistrer le modèle affiné dans un dossier spécifique
* **Utilité** : Permettre une réutilisation future sans devoir refaire tout l’entraînement

---

### **Étape 6 – Évaluation du modèle**

* **Test** : Poser plusieurs questions issues du domaine sur lequel le modèle a été affiné, comparer les réponses à celles du modèle d’origine
* **Indicateurs** : Précision, exactitude de l’extraction, capacité à pointer la bonne séquence dans le texte
* **Comparaison** : (optionnel) Faire une évaluation objective sur un lot de questions/tests avant et après fine-tuning

---

### **Étape 7 – (Optionnel) Création de son propre dataset**

* **But** : Aller plus loin que le SQuAD généraliste en créant un corpus spécifique (docs métier, articles, FAQ interne…)
* **Action** : Structurer les données selon le format SQuAD, répéter le processus d’affinage

---

### **Étape 8 – (Optionnel) Pipeline complet avec récupérateur**

* **Objectif** : Associer le module de “retrieval” (recherche documentaire) pour créer un pipeline complet RAG (retrieval-augmented generation)
* **Gain** : Permettre au système d’aller chercher dans de larges bases documentaires la partie de texte pertinente avant l’extraction de la réponse

---

## 3. **Robustesse de l’approche**

* **Simplicité** : Utilisation d’outils standards, structuration claire étape par étape
* **Modularité** : Chacune des étapes est indépendante et réutilisable
* **Évolutivité** : Possibilité d’améliorer facilement le modèle en ajoutant des données ou en affinant d’autres modèles
* **Portabilité** : Fonctionne en local ou sur cloud, peut être facilement partagé (Colab)

---

## 4. **Points de vigilance**

* **Formatage exact du dataset** (clé d’un fine-tuning réussi)
* **Capacité GPU** : Ajuster les paramètres pour éviter la saturation ou l’overfitting
* **Évaluation objective** : Toujours comparer les performances avant/après pour mesurer le gain réel

# Cellule 2 : Vérification du GPU et configuration de la télémétrie (optionnel)

# Cellule 3 : Téléchargement et préparation d’un dataset SQuAD

```python
# 3. Télécharger un jeu de données SQuAD (ou charger le vôtre)
# Exemple avec SQuAD v2.0 officiel
import requests, os

os.makedirs("data", exist_ok=True)
url = "https://rajpurkar.github.io/SQuAD-explorer/dataset/train-v2.0.json"
local_path = "data/train-v2.0.json"

if not os.path.exists(local_path):
    r = requests.get(url)
    with open(local_path, "wb") as f:
        f.write(r.content)
    print(f"Téléchargé : {local_path}")
else:
    print("Dataset déjà présent :", local_path)
```

# Cellule 4 : Initialisation du FARMReader pré-entraîné

```python
# 4. Charger le FARMReader pré-entraîné pour SQuAD2
from haystack.nodes import FARMReader

model_name = "deepset/roberta-base-squad2"
reader = FARMReader(model_name_or_path=model_name, use_gpu=torch.cuda.is_available())
print("FARMReader initialisé avec le modèle :", model_name)
```

# Cellule 5 : Fine-tuning du modèle sur vos données

```python
# 5. Affiner le modèle sur un dataset SQuAD customisé ou officiel
# Attention : pour l'exemple, on s'arrête après 1 epoch (à adapter à vos besoins réels !)

data_dir = "data"
train_filename = "train-v2.0.json"  # À remplacer par le nom de votre dataset custom le cas échéant

reader.train(
    data_dir=data_dir,
    train_filename=train_filename,
    use_gpu=torch.cuda.is_available(),
    n_epochs=1,            # À augmenter pour un vrai fine-tuning
    save_dir="my_reader_model",
    batch_size=4,          # Ajustez selon la RAM GPU disponible
    evaluate_every=1000,   # Pour SQuAD2 officiel (long !)
    learning_rate=3e-5,
)

print("Entraînement terminé. Modèle sauvegardé dans le dossier my_reader_model.")
```

# Cellule 6 : Chargement du modèle affiné pour utilisation

```python
# 6. Charger le modèle affiné pour l'inférence
reader_finetuned = FARMReader(model_name_or_path="my_reader_model", use_gpu=torch.cuda.is_available())
print("Modèle affiné chargé depuis my_reader_model.")
```

# Cellule 7 : Tester le modèle affiné (question-réponse simple)

```python
# 7. Tester le modèle avec une question sur un contexte choisi
context = """
Haystack est un framework open-source pour le traitement du langage naturel développé par deepset. Il permet de créer des systèmes de recherche et de questions-réponses puissants à partir de grands modèles de langage et de pipelines modulaires.
"""

question = "Qui a développé Haystack ?"

results = reader_finetuned.predict_on_texts(
    question=question,
    texts=[context]
)

print("Question :", question)
print("Réponse extraite :", results['answers'][0].answer)
print("Confiance :", results['answers'][0].score)
print("Position de la réponse dans le texte :", results['answers'][0].offsets_in_context)
```

# Cellule 8 : Évaluation et comparaison (optionnel)

```python
# 8. (Optionnel) Comparer le modèle avant/après fine-tuning
# Charger le modèle d'origine
reader_base = FARMReader(model_name_or_path="deepset/roberta-base-squad2", use_gpu=torch.cuda.is_available())

# Prédiction avant fine-tuning
result_base = reader_base.predict_on_texts(question=question, texts=[context])
print("[Avant fine-tuning] Réponse :", result_base['answers'][0].answer)

# Prédiction après fine-tuning
result_ft = reader_finetuned.predict_on_texts(question=question, texts=[context])
print("[Après fine-tuning] Réponse :", result_ft['answers'][0].answer)
```

```python
### Comparaison modèle base vs modèle fine-tuné sur une question simple

context = """
Haystack est un framework open-source pour le traitement du langage naturel développé par deepset.
Il permet de créer des systèmes de recherche et de questions-réponses puissants à partir de grands modèles de langage et de pipelines modulaires.
"""

question = "Qui a développé Haystack ?"

# Prédiction avec le modèle d'origine (avant fine-tuning)
result_base = reader_base.predict_on_texts(question=question, texts=[context])
answer_base = result_base['answers'][0].answer
score_base = result_base['answers'][0].score

# Prédiction avec le modèle fine-tuné
result_ft = reader_finetuned.predict_on_texts(question=question, texts=[context])
answer_ft = result_ft['answers'][0].answer
score_ft = result_ft['answers'][0].score

print("---- Évaluation comparative ----")
print(f"Question : {question}\n")
print("[Avant fine-tuning]")
print(f"Réponse extraite : {answer_base}")
print(f"Score de confiance : {score_base:.3f}")

print("\n[Après fine-tuning]")
print(f"Réponse extraite : {answer_ft}")
print(f"Score de confiance : {score_ft:.3f}")
```
