# 🌟 Exercice 1 : Réécrire et optimiser une invite vague
Scénario :
Vous êtes ingénieur en IA dans une start-up spécialisée dans la productivité. Votre responsable vous demande d'utiliser ChatGPT pour créer une publication LinkedIn promouvant sa nouvelle application de concentration, FlowNest . Il vous envoie cette vague invite :

« Écrivez quelque chose sur les conseils de productivité. »



Tâches
Identifiez trois problèmes critiques avec cette invite.
(Indice : pensez à la voix de la marque, aux utilisateurs cibles et à l’engagement sur la plateforme.)

Réécrivez l'invite en utilisant les trois techniques :

Incitation à jouer un rôle (par exemple, « Agir en tant que… »)
Instructions basées sur des instructions (par exemple, « Donnez 3 actions réalisables… »)
Contrainte de format/ton/longueur (par exemple, « Utiliser des puces » ou « 280 caractères maximum »)

Ajoutez une dernière ligne qui rend l’invite réutilisable (par exemple, « Utilisez ce format chaque fois que nous faisons la promotion d’outils de productivité sur les réseaux sociaux »).

# Résolution complète et pas-à-pas de l’exercice 1

Je vais procéder en 3 étapes claires et exécutables : (A) diagnostic des problèmes critiques, (B) réécriture de l’invite en appliquant les 3 techniques demandées, (C) justification et exemple de sortie (vérifiable).

---

## A — Trois problèmes critiques avec l’invite initiale

Invite initiale : « Écrivez quelque chose sur les conseils de productivité. »

1. **Absence d’audience et d’objectif**
   — Problème : on ne sait pas pour qui c’est (professionnels, étudiants, débutants?) ni quel objectif (informer, vendre, générer une démo).
   — Impact : le ton, le niveau de détail et l’appel à l’action seront inappropriés ou inefficaces.

2. **Pas de format ni de contrainte**
   — Problème : aucune indication sur la longueur, le format (post LinkedIn, fil Twitter, article) ni sur la structure.
   — Impact : sortie trop longue, pas adaptée au canal, peu engageante (ex. un texte long pour une plateforme où l’engagement requis est bref).

3. **Absence de voix de marque et d’éléments actionnables**
   — Problème : la consigne ne précise pas la voix (crédible, ludique, technique) ni d’actions concrètes liées au produit FlowNest.
   — Impact : résultat générique, non différenciant, qui ne met pas en valeur l’application.

---

## B — Réécriture de l’invite (intégrant les 3 techniques demandées)

Voici **une** invite complète qui applique : (1) incitation par rôle, (2) instructions basées sur actions, (3) contraintes de format/ton/longueur. La dernière ligne rend l’invite réutilisable.

> **Invite optimisée (à envoyer à ChatGPT)**
> Agissez en tant que *Responsable Marketing de FlowNest*. Public cible : professionnels urbains 25–45 ans cherchant à améliorer leur concentration au travail. Objectif : susciter l’intérêt et obtenir des inscriptions à la démo.
> Tâches précises :
>
> 1. Rédigez **une publication LinkedIn courte (≤ 280 caractères)** comprenant : une accroche percutante (1 phrase), **3 conseils pratiques** directement liés à l’utilisation de FlowNest (liste de 3 puces courtes), et un **appel à l’action clair** (ex. tester la démo).
> 2. Style/voix : professionnel, bienveillant, orienté résultat — langage simple et crédible.
>    Contraintes : n’inventez pas de chiffres non vérifiés ; si vous mentionnez des résultats, précisez « tests internes » ou « bêta » ; utilisez des puces ; respectez la limite de 280 caractères (espaces inclus).
>    **Réutilisation :** Utilisez ce format chaque fois que nous faisons la promotion d’outils de productivité sur les réseaux sociaux.

---

## C — Justification des choix + Exemple de sortie

### Pourquoi cette invite évite les problèmes identifiés

1. **Audience & objectif explicites** → le modèle adapte le ton et le CTA.
2. **Format et contrainte de longueur** → garantit une sortie adaptée à LinkedIn (post court, lisible).
3. **Rôle + instructions actionnables** → produit des conseils connectés au produit (utile pour conversion), pas un texte générique.

### Exemple de publication générée (respecte la contrainte ≤ 280 caractères)

Voici un exemple que le prompt ci-dessus devrait produire. J’ai vérifié le nombre de caractères.

Texte généré :
FlowNest : retrouvez votre concentration. 3 actions concrètes — 1) sessions 25 min guidées; 2) mode silence intelligent; 3) suivi simple des progrès. Testez la démo gratuite : \[lien]

Nombre de caractères (espaces inclus) : **181** — donc < 280.

### Remarques pratiques pour l’utilisation

* Si l’équipe marketing veut **mettre une preuve sociale**, adapter l’invite : « Si disponible, ajoutez 1 preuve sociale brève (ex. “+15 % de concentration en bêta”), en précisant la source. »
* Pour des visuels LinkedIn, ajouter : « Fournissez aussi 1 suggestion de visuel (texte court pour l’image). »
* Pour réutilisabilité multi-canal, conservez la même structure mais changez la contrainte de longueur (ex. 280 → 140 pour Twitter/X, ou \~150–250 mots pour article LinkedIn).

# Exercice 2 : Questionnaire en plusieurs parties pour la génération de questionnaires
Scénario :
Un professeur de sciences de 5e publie un court article sur les éruptions volcaniques . Vous devez rédiger une invite pour ChatGPT afin de créer du contenu pédagogique captivant.



Tâches
Rédigez une invite complète qui demande au modèle de :
Résumez l'article en 2 points simples
Générez 3 questions à choix multiples adaptées à l'âge , chacune avec 1 bonne réponse et 2 distracteurs
Inclure un ton amical et un vocabulaire simple pour les 11-13 ans
Affichez le tout dans un format adapté à Google Slides

Comparez votre invite à celle-ci, vague :

« Créez un quiz pour les enfants sur cet article. »

Expliquez trois manières clés par lesquelles votre version évite les résultats non pertinents ou de mauvaise qualité.

(A) une invite prête à copier/coller pour ChatGPT, (B) une comparaison avec l’invite vague, et (C) un exemple de sortie formatée pour Google Slides (modèle) afin que vous puissiez vérifier le résultat attendu.

# A — Invite complète (à copier/coller dans ChatGPT)

> Agis en tant qu’enseignant·e de sciences pour des élèves de 5ᵉ (11–13 ans). Tu recevras ci-dessous le texte d’un court article sur les éruptions volcaniques.
> **Consignes strictes :**
>
> 1. Résume l’article en **2 points simples**, chacun en une phrase courte (vocabulaire simple, compréhensible par des 11–13 ans).
> 2. Génère **3 questions à choix multiples** adaptées à cet âge. Pour chaque question :
>
>    * Donne **l’énoncé** de la question (clair et court).
>    * Propose **3 choix** nommés A, B, C.
>    * Indique **la bonne réponse** (lettre) et une **brève explication** (1 phrase) destinée à l’enseignant.
> 3. Le **ton** doit être amical, encourageant et simple. Évite les phrases longues et le vocabulaire scientifique complexe.
> 4. **Format pour Google Slides** : fournis la sortie sous forme de diapositives séparées par une ligne contenant `---` (trois tirets). Pour chaque diapositive, commence par `Titre :` puis `Contenu :` et utilise des puces si nécessaire. Exemple de structure :
>
>    ```
>    Titre : Titre de la diapositive
>    Contenu :
>    - point 1
>    - point 2
>    ```
> 5. Ajoute **une diapositive finale** intitulée `Réponses (enseignant)` qui liste, pour chaque question, la bonne réponse et l’explication.
> 6. **Ne complète rien sans l’article** : si l’article n’est pas fourni, répond seulement : « Article manquant — veuillez coller le texte. »
>    **Remarque** : ne crée pas d’images, ne cite pas de sources externes, et n’invente pas d’informations qui ne figurent pas dans l’article.

# B — Pourquoi cette invite est supérieure à « Créez un quiz pour les enfants sur cet article. »

1. **Précise l’audience et le niveau** — en indiquant « 5ᵉ (11–13 ans) » on force un vocabulaire et une complexité adaptés ; la version vague peut produire des questions trop techniques ou trop simples.
2. **Structure et format explicite (Google Slides)** — la consigne `---` + `Titre : / Contenu :` évite les sorties non structurées et facilite le collage direct dans Slides ; la version vague ne garantit pas le format attendu.
3. **Contraintes et métadonnées pédagogiques** — demander la bonne réponse + courte explication destinée à l’enseignant réduit les risques d’hallucinations et permet une correction rapide ; la consigne vague laisse le modèle libre de ne pas fournir de clé ou d’explications.

# C — Exemple de sortie attendue (format Google Slides)

> *Remarque : ceci est un exemple type. Remplacez par le contenu réel produit après insertion de l’article.*

```
Titre : Éruptions volcaniques - Résumé
Contenu :
- Les volcans émettent de la lave, des cendres et des gaz lors d’une éruption.
- Les éruptions peuvent être explosives (nuées, cendres) ou effusives (cours de lave) et présentent des risques pour les personnes et les bâtiments.

---
Titre : Question 1
Contenu :
- Question : Qu’est-ce qu’un volcan émet principalement pendant une éruption ?
  A) De la glace
  B) De la lave, des cendres et des gaz
  C) Du sable
- Indice : relire le premier point du résumé.

---
Titre : Question 2
Contenu :
- Question : Quelle est la différence principale entre une éruption explosive et une éruption effusive ?
  A) Explosive = beaucoup de cendres et nuées ; effusive = coulées de lave lentes
  B) Explosive = seulement de la fumée ; effusive = seulement de l’eau
  C) Explosive = vent fort ; effusive = vent faible
- Indice : relire le deuxième point du résumé.

---
Titre : Question 3
Contenu :
- Question : Quelle mesure permet de réduire le risque pour les personnes près d’un volcan ?
  A) S’éloigner et suivre les consignes des autorités
  B) Rester à l’entrée du volcan pour observer
  C) Ramasser des cendres pour les garder chez soi
- Indice : penser aux risques pour les populations.

---
Titre : Réponses (enseignant)
Contenu :
- Question 1 : Réponse B. Explication : les volcans expulsent principalement de la lave, des cendres et des gaz lors d’une éruption.
- Question 2 : Réponse A. Explication : une éruption explosive projette beaucoup de cendres et formes de nuées ardentes ; une éruption effusive produit surtout des coulées de lave.
- Question 3 : Réponse A. Explication : la sécurité passe par l’éloignement et le respect des consignes officielles ; rester près du volcan est dangereux.
```

> 1. Fournissez **exactement 3 points clés** (numérotés 1→3). Pour chaque point : une phrase synthétique + **au moins 1 chiffre** (montant en € ou %), la **variation mois-sur-mois (M/M)** et la **variation vs budget** si disponible. Exemple : « CA: 2,1 M€ (+3 % M/M; -1 % vs budget) ».
> 2. Ajoutez **une recommandation concrète** (1 phrase, action prioritaire).
> 3. Ajoutez **les 3 principaux risques/opportunités** en une ligne chacun (bullet).
>    **Format de sortie (obligatoire) :**
>
> * **Partie A — Slide (texte prêt)** : Titre, puis 3 bullets (les points clés), puis une ligne « Recommandation ». Limitez la slide à **\~150–220 mots**.
> * **Partie B — Notes du présentateur** : 3 courtes lignes de script pour chaque point (indiquez le **temps recommandé** en secondes ; ex. 40s), et une phrase d’ouverture et de clôture. Durée totale : \~3 minutes.
>   **Contraintes supplémentaires :**
> * **Ne pas inventer de chiffres.** Si une donnée n’apparaît pas dans la source, écrivez « Donnée non fournie ».
> * Utilisez un **style concis, non technique, orienté décision**. Phrases courtes.
> * Renseignez les unités (€ ou %) entre parenthèses et signalez la période (mois concerné).
>   **Sortie attendue :** renvoyez uniquement le texte des parties A et B, en clair, sans ajout d’analyse externe ou liens.

# Exercice 4 : Associer l’invite à l’objectif
Scénario :

Vous créez des modèles d'invite pour un chatbot de support client pour une entreprise de e-commerce. Choisissez l'un des styles suivants et utilisez-le pour rédiger une invite complète pour le chatbot.

Exploratoire : aide le modèle à réfléchir à des options ou à générer des idées
Structuré : suit un format fixe (listes, sections, étapes)
Conversationnel : imite un ton humain et bavard
Fonctionnel : Exécute une tâche avec précision (par exemple, résumer ou transformer une entrée)


Tâches
Écrivez votre invite en utilisant le style sélectionné.
Identifiez 2 caractéristiques clés qui font que votre invite correspond au style.
Justifiez votre choix : pourquoi ce style est-il meilleur que les autres pour votre cas spécifique ?

Je choisis le style **Structuré** — le plus adapté pour un chatbot de support client e-commerce qui doit répondre de façon fiable, rapide et homogène.

# Invite complète (style structuré) — à coller comme system prompt / prompt principal

Agis en tant que **chatbot de support client structuré** pour la boutique e-commerce **\[NomEntreprise]** (site web, chat en ligne). Tu dois résoudre les demandes clients de façon claire, concise et traçable. Respecte strictement le format de sortie et les règles ci-dessous.

**Contexte**

* Canal : chat textuel sur le site.
* Langue : français (si client écrit en autre langue, répondre en français sauf indication contraire).
* Données clients : si l’utilisateur fournit un numéro de commande, utilise-le dans la réponse. Si non, demande-le.

**Objectifs**

1. Comprendre rapidement la demande.
2. Proposer une solution concrète ou une action suivante clairement libellée.
3. Si nécessaire, récupérer les informations manquantes ou escalader.

**Format de sortie (obligatoire)** — renvoie **uniquement** ce bloc structuré (pas de texte libre avant) :

```
[Résumé du problème] : <1 phrase très courte résumant la demande>
[Action proposée] :
- Étape 1 : <action immédiate / information fournie>
- Étape 2 : <action client requise si applicable>
- Étape 3 : <prochaine étape / délai>
[Informations manquantes] : <liste ou "Aucune">
[Escalade nécessaire] : Oui / Non — <si oui : motif précis, ex. "litige paiement", "demande légale", "client agressif">
[Message final au client] : <1 phrase empathique + CTA clair>
```

**Règles opérationnelles**

* Priorise la sécurité et la précision ; ne devine **aucun** chiffre de commande, prix ou statut si l’info n’est pas fournie : écris « Donnée non fournie » dans \[Informations manquantes].
* Si la demande concerne : remboursement avec preuve, litige de paiement, menace, ou question légale, **mettre \[Escalade nécessaire] : Oui** et indiquer « Transmettre au support humain — équipe Litiges ». Ne proposer pas d’action définitive (ex : remboursement) si le système n’a pas accès aux outils.
* Pour demandes standards (suivi commande, retour, annulation, info produit) : fournir procédure standard et délais (ex. « étiquette de retour envoyée sous 24 h ouvrées » — uniquement si c’est une règle opérationnelle réelle ; sinon indiquer « délai estimé »).
* Longueur : **max. 6 lignes** de contenu total (toute la structure).
* Ton : professionnel, empathique, factuel. Éviter jargon technique.

**Questions à poser si informations manquantes (templates)**

* « Pouvez-vous me donner votre numéro de commande ? (ex : #123456) »
* « Quelle est la référence / couleur / taille du produit concerné ? »
* « Souhaitez-vous un échange ou un remboursement ? »

**Exemples d’escalade (motifs précis)**

* Paiement contesté/chargeback, demande juridique, menace, demande d’annulation après expédition irréversible, problème qualité causant risque de sécurité.

**Comportement interdit**

* Ne jamais inventer de délais, montants ou confirmations d’actions opérationnelles si tu n’as pas l’accès système.
* Ne pas donner de conseils juridiques ou médicaux.

---

# 2 caractéristiques clés qui rendent l’invite « structuré »

1. **Format de sortie strict et reproductible** : la réponse doit suivre un bloc contenant des champs nommés (\[Résumé du problème], \[Action proposée], etc.). Cela garantit cohérence, facilite l’audit et l’intégration (logs, transfert humain).
2. **Checklists et templates intégrés** : questions à poser, règles d’escalade et comportements interdits. Ces éléments transforment une réponse libre en une procédure standardisée, réduisant les erreurs humaines et les hallucinations.

# 3 — Justification : pourquoi le style structuré est meilleur ici

* Le support e-commerce doit fournir des réponses **précises, actionnables et traçables** (ex : statut de commande, procédure de retour). Le style structuré force le bot à produire des sorties normalisées exploitables par des agents humains ou des systèmes back-office.
* Comparé aux autres styles :

  * **Exploratoire** risquerait des réponses trop ouvertes et non immédiatement exploitables.
  * **Conversationnel** peut améliorer l’empathie mais fait perdre en clarté opérationnelle et compliquer l’escalade.
  * **Fonctionnel** (exécution précise) est utile mais dépend d’intégrations techniques ; le style structuré est un meilleur compromis car il définit des actions claires tout en restant indépendant des accès systèmes.

# 🌟 Exercice 6 : Détection et atténuation des hallucinations
Scénario :
Vous préparez le résumé d'un article évalué par des pairs sur le changement climatique et la biodiversité marine . ChatGPT insère cette affirmation :

« Plus de 50 % des espèces marines devraient disparaître d’ici 2050. »

Cette affirmation n’apparaît pas dans l’article original.



Tâches
1. Rédigez une invite précise demandant au modèle d'éviter les hallucinations et d'utiliser uniquement le texte fourni .

2. Réécrivez l'invite en utilisant un langage de vérification , comme :

« N'incluez que les affirmations explicitement énoncées dans l'article. »
« En cas de doute, indiquez que l’information n’est pas disponible. »
3. Énumérez 2 stratégies d’atténuation que vous avez utilisées pour réduire le risque d’hallucinations.

4. Nommez 2 domaines professionnels (par exemple, juridique, santé) dans lesquels les hallucinations sont particulièrement risquées et expliquez pourquoi.

## 1) Invite stricte demandant d’éviter les hallucinations (à coller dans l’IA)

> Agis en tant qu’**analyste scientifique**. Utilise **UNIQUEMENT** le texte fourni ci-dessous. **Ne fais aucune affirmation** qui ne figure pas explicitement dans ce texte. Pour chaque affirmation que tu produis, fournis obligatoirement :
>
> 1. **Citation exacte** (≤ 25 mots) extraite du texte entre guillemets.
> 2. **Localisation** de la citation (page / paragraphe / ligne).
> 3. **Statut** : EXPLICITE / INFÉRÉE / NON FOURNIE.
> 4. **Courte justification** (1 phrase) expliquant le lien entre la citation et l’affirmation.
>    Si une donnée chiffrée n'apparaît pas dans le texte, écris « **Information non fournie** ». **Ne génère jamais de chiffres, dates ou pourcentages non présents dans le texte.**
>    **Format de sortie (obligatoire)** : table Markdown avec les colonnes
>    `Affirmation | Citation (≤25 mots) | Localisation | Statut | Justification`
>    Si l’article n’est pas fourni, renvoie uniquement : « Article manquant — veuillez coller le texte. »

---

## 2) Invite réécrite avec langage de vérification (phrases à intégrer)

Intègre explicitement ces phrases dans le prompt (exemples) :

* **« N'incluez que les affirmations explicitement énoncées dans l'article. »**
* **« En cas de doute, indiquez que l’information n’est pas disponible (écrivez exactement : “Information non fournie”). »**
* **« Toute affirmation quantitative doit être accompagnée de la citation textuelle exacte et de sa localisation ; sinon marquez-la comme “Information non fournie”. »**

Version complète (prête à copier) — combine les deux éléments :

> Agis en tant qu’analyste scientifique. N’incluez **que** les affirmations explicitement énoncées dans l’article fourni. En cas de doute, indiquez « Information non fournie ». Pour chaque affirmation, fournissez : (a) citation exacte ≤25 mots entre guillemets, (b) localisation (page/para/ligne), (c) statut EXPLICITE / INFÉRÉE / NON FOURNIE, (d) justification 1 phrase. Si une donnée chiffrée n’apparaît pas, écrivez « Information non fournie ». Sortie : table Markdown `Affirmation | Citation | Localisation | Statut | Justification`. Si article manquant : répondre « Article manquant — veuillez coller le texte. »

---

## 3) Deux stratégies d’atténuation utilisées (et pourquoi elles fonctionnent)

1. **Exiger preuve textuelle (quote + localisation + statut)**

   * Description : toute affirmation doit être appuyée par une citation textuelle extraite du document (≤25 mots) et indiquer où elle se trouve.
   * Pourquoi cela réduit les hallucinations : contraint le modèle à lier chaque assertion à une preuve concrète dans la source. Sans preuve explicite, l’affirmation est marquée « NON FOURNIE » et n’est pas présentée comme un fait. Cela bloque la production de chiffres inventés.

2. **Politique conservative « Information non fournie » + statut explicite INFÉRÉE vs EXPLICITE**

   * Description : imposer une réponse par défaut « Information non fournie » pour toute donnée absente ; distinguer les éléments explicitement présents et ceux qui sont des inférences (avec label clair).
   * Pourquoi cela réduit les hallucinations : force la transparence sur le degré de certitude et interdit au modèle de transformer une inférence en fait. L’étiquetage explicite aide l’utilisateur humain à filtrer ce qui nécessite vérification externe.

---

## 4) Deux domaines professionnels où les hallucinations sont particulièrement risquées — et pourquoi

1. **Santé / Médecine**

   * Pourquoi risqué : une affirmation incorrecte (diagnostic, posologie, effet secondaire) peut conduire à un traitement inapproprié, mettre en danger la vie du patient ou provoquer des erreurs médicales. Les conséquences sont directes et potentiellement graves.

2. **Droit / Juridique**

   * Pourquoi risqué : une information juridique inventée (dates légales, obligations contractuelles, montants, interprétations de loi) peut entraîner des décisions litigieuses, des pertes financières, ou engager la responsabilité des professionnels. Les erreurs peuvent avoir des conséquences légales et financières lourdes.

(À titre secondaire : Finance/investissement est aussi critique — fausses données chiffrées peuvent induire des pertes.)

---

## Checklist rapide que vous pouvez coller dans le prompt (version compacte)

* Utiliser uniquement le texte fourni.
* Pour chaque affirmation : citation ≤25 mots + localisation + statut (EXPLICITE / INFÉRÉE / NON FOURNIE) + 1 phrase de justification.
* Si chiffre absent : écrire « Information non fournie ».
* Sortie = table Markdown `Affirmation | Citation | Localisation | Statut | Justification`.

# Projets innovants, creatifs et concrtes potentiels

# Projet 1 — Summarizer scientifique « citation-first »

**Objectif**
Produire des résumés exécutifs de publications scientifiques strictement basés sur le texte source, avec citation obligatoire et localisation pour chaque affirmation.

**Description**
Un service (API + UI) qui prend un PDF ou texte d’article et génère automatiquement une table « Affirmation | Citation | Localisation | Statut | Justification » suivant le prompt vérifié. Sortie exportable en Markdown/CSV et slide prête à présenter. Interface pour l’éditeur humain où on peut valider/modifier les citations.

**Composants techniques**

* Ingestion PDF -> OCR (PyMuPDF / pdfminer + OCR si nécessaire)
* Paragraphe/segmentation, indexation (FAISS/Weaviate)
* Pipeline RAG contrôlé (LangChain/Haystack) avec prompt strict « quote+localisation+statut »
* UI web (React) pour revue et correction humaine
* Backend (FastAPI), DB (Postgres), containerisation Docker

**Livrables MVP (6–8 semaines)**

* API ingest + extraction textuelle
* Module RAG + prompt vérifié, génération table Markdown
* UI simple pour revue + export CSV/Slides
* Tests sur 20 articles de validation

**Métriques de succès**

* % d’affirmations accompagnées d’une citation valide (>95%)
* Réduction des hallucinations mesurée vs résumé baseline (target >80% réduction)
* Taux d’acceptation éditeur humain (>85%)

**Risques & mitigations**

* OCR erroné → checks de qualité OCR + signaler « Information non fournie » si faible confiance.
* Modèle qui paraphrase sans citer → forcer extraction span via retriever et template strict.

# Projet 2 — CI/CD « Hallucination Gate » pour déploiement LLM en entreprise

**Description**
Pipeline d’intégration qui exécute un jeu de tests (assertions fact/quote tests, contrainte « info non fournie ») sur chaque nouvelle version de prompt ou modèle avant déploiement. Génère rapport de conformité (pass/fail) et métriques historiques.

**Composants techniques**

* Test harness (pytest-like) avec scénarios de test (inputs + expected behavior)
* Orchestrateur CI (GitHub Actions/GitLab CI)
* Comparateur d’assertions (string match / fuzzy match) et heuristiques de confiance
* Dashboard de monitoring (Grafana/Metabase)

**Livrables MVP (4–6 semaines)**

* Kit de tests réutilisable (templates)
* Intégration GitHub Actions + exemples de jobs
* Dashboard de résultats + alerting

**Métriques de succès**

* % builds bloqués pour non-conformité (indicateur de l’efficacité des tests)
* Diminution des incidents en prod liés à hallucinations (suivi post-prod)

**Risques & mitigations**

* Faux positifs bloquant release → permettre override humain après justification et amélioration des tests.
* Scalabilité des tests → prioriser tests critiques (sanity checks) puis élargir.
