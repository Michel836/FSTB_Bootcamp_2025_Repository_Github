````python
# Installer (exécuter dans un terminal avant cette cellule) :
# pip install -U "numpy<2" langchain faiss-cpu sentence-transformers

from langchain.embeddings.base import Embeddings
from sentence_transformers import SentenceTransformer
from langchain.vectorstores import FAISS
from langchain.schema import Document
import numpy as np
from typing import List

# 🚀 Classe d'embedding personnalisée
class MyEmbeddings(Embeddings):
    def __init__(self, model_name: str = "sentence-transformers/all-MiniLM-L6-v2"):
        self.model = SentenceTransformer(model_name, device="cpu")
    def embed_documents(self, texts: List[str]) -> List[List[float]]:
        return self.model.encode(texts, normalize_embeddings=True).tolist()
    def embed_query(self, text: str) -> List[float]:
        return self.model.encode([text], normalize_embeddings=True)[0].tolist()

# 📄 Charger les documents (texte simple ou loader PDF/TXT)
texts = [
    "Le climat change et impacte les écosystèmes.",
    "L'IA transforme les processus industriels.",
    "Les véhicules électriques révolutionnent les transports."
]
docs = [Document(page_content=t) for t in texts]

# 🧠 Initialiser les embeddings
emb = MyEmbeddings()

# ✅ Créer le vector store avec FAISS.from_documents
vector_store = FAISS.from_documents(
    documents=docs,
    embedding=emb,
)

# 🔍 Recherche par similarité
query = "Comment l'intelligence artificielle impacte l'industrie ?"
results = vector_store.similarity_search(query, k=2)
for res in results:
    print("➡️", res.page_content)
````

````python
# 🔧 Installer avant d’exécuter :
# pip install -U "numpy<2" langchain sentence-transformers faiss-cpu transformers accel

from sentence_transformers import SentenceTransformer
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
from langchain.embeddings.base import Embeddings
from langchain.vectorstores import FAISS
from langchain.schema import Document
from langchain.chains import RetrievalQA
from langchain.llms import HuggingFacePipeline
import numpy as np
from typing import List

# 📦 Classe embeddings locale
class MyEmbeddings(Embeddings):
    def __init__(self, model_name="sentence-transformers/all-MiniLM-L6-v2"):
        self.model = SentenceTransformer(model_name, device="cpu")
    def embed_documents(self, texts: List[str]) -> List[List[float]]:
        return self.model.encode(texts, normalize_embeddings=True).tolist()
    def embed_query(self, text: str) -> List[float]:
        return self.model.encode([text], normalize_embeddings=True)[0].tolist()

# 🛠️ 1. Charger vos docs
texts = [
    "La LLM peut générer du texte contextualisé.",
    "Le fine‑tuning ajuste les poids du modèle sur des données spécifiques.",
]
docs = [Document(page_content=t) for t in texts]

# 2. embeddings
emb = MyEmbeddings()
vector_store = FAISS.from_documents(documents=docs, embedding=emb)

# 3. Initialiser TinyLlama via HuggingFacePipeline
tokenizer = AutoTokenizer.from_pretrained("TinyLlama/TinyLlama-1.1B-Chat-v1.0")
model = AutoModelForCausalLM.from_pretrained(
    "TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    device_map="cpu",
    torch_dtype="auto"
)
gen = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=256,
    temperature=0.7,
    top_p=0.9
)
llm = HuggingFacePipeline(pipeline=gen)

# 4. Construire pipeline RAG
qa = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=vector_store.as_retriever(search_kwargs={"k":3}),
    return_source_documents=True
)

# 5. Tester
question = "Qu'est-ce que le fine‑tuning ?"
result = qa({"query": question})
print("\n🧠 Réponse :\n", result['result'])
print("\n📚 Sources récupérées :")
for d in result['source_documents']:
    print("-", d.page_content)
````
