### 1. Qu’est-ce que la tokenisation ?

La tokenisation consiste à découper un texte en unités (tokens) – mots, sous-mots, caractères ou signes de ponctuation –, puis à convertir chaque token en un identifiant (entier) via un dictionnaire (vocabulaire). Les LLMs utilisent souvent la Byte-Pair Encoding (BPE), qui combine des sous-mots fréquents pour créer des tokens efficaces .
Ce qui permet au modèle de traiter le texte comme une suite de nombres au lieu de chaînes de caractères.

### 2. Exemple de code

```python
from transformers import AutoTokenizer
import matplotlib.pyplot as plt

# 1. Texte choisi
text = "La tokenisation découpe du texte en petits morceaux."

# 2. Chargement du tokenizer
tokenizer = AutoTokenizer.from_pretrained("gpt2")

# 3. Tokenisation
tokens = tokenizer.tokenize(text)
token_ids = tokenizer.convert_tokens_to_ids(tokens)

# 4. Affichage
print(f"Texte original : {text}")
print(f"Tokens         : {tokens}")
print(f"Token IDs      : {token_ids}")

# 5. Visualisation
plt.figure(figsize=(10, 4))
plt.bar(tokens, token_ids, color="skyblue")
plt.xlabel("Tokens")
plt.ylabel("ID numériques")
plt.title("Visualisation de la tokenisation")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

3. Interprétation

* **Token “Ġtex” domine la barre** avec un ID très élevé (\~49 000), car il correspond à un sous‑mot rare dans le vocabulaire GPT‑2 et illustre la granularité byte‑level BPE qui met l’accent sur des segments moins fréquents.
* **“GdÃ©” (ou “ĠdÃ©” selon l’étiquette)** apparaît avec un ID élevé (\~39 000) : c’est le symptôme du mojibake (“découpe” mal encodé), ce token spécifique n’existe pas dans un vocabulaire bien formé.
* **Tokens fréquents** comme “La” (ID 14772), “Ġdu” (7043) ou “.” (13) ont des IDs plus faibles, reflétant leur occurrence régulière dans le corpus d’entraînement.
* **Fragments courts** (“c”, “ou”, “pe”, “its”, “ce”, “aux”) ont aussi des IDs bas à modérés, car ils représentent des sous‑mots courants, tandis que la segmentation de “tokenisation” en plusieurs morceaux montre l’aspect sous‑mot BPE.
* **Interprétation** :

  * Un ID élevé = token rare ou mal formé (souvent conséquence de problèmes d’encodage).
  * Un ID bas ou moyen = token fréquent, attestant d’une segmentation plus naturelle.
* **Conclusion** : la visualisation met en évidence à la fois la répartition des IDs (fréquence vs rareté) et l’impact néfaste d’un mauvais encodage UTF‑8 sur la qualité des tokens.

### 1. Afficher l’ID de chaque token

```python
from transformers import AutoTokenizer

text = "Les LLMs sont fascinants et puissants."

tokenizer = AutoTokenizer.from_pretrained("gpt2")

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print(f"Texte original : {text}")
print(f"Tokens         : {tokens}")
print(f"IDs des tokens : {token_ids}")

# Afficher token par token avec son ID
for tok, tid in zip(tokens, token_ids):
    print(f"{tok} → {tid}")
```

Le tokenizer GPT‑2 segmente “Les LLMs sont fascinants et puissants.” en 13 sous‑mots de type byte‑level BPE, chacun mappé à un identifiant unique. Le préfixe `Ġ` marque chaque début de mot (boundary) et les fragments (`fasc`+`in`+`ants`) illustrent la décomposition des mots plus longs en sous‑unités fréquentes.

---

### 1. Rôle du préfixe `Ġ`

* Le caractère **`Ġ`** indique un **espace** précédant le token dans le texte original, marquant le début d’un nouveau mot ([Medium][1]).
* Par exemple, `ĠLL` signifie que “LL” commence après un espace, tandis que `Ms` sans préfixe est collé au token précédent.

---

### 2. Découpage en sous‑mots (BPE)

* **“Les”** (ID 35882) et **“ont”** (ID 756) sont des tokens entiers, fréquemment vus lors de l’entraînement.
* **“LLMs”** a été découpé en `ĠLL` (27140) + `Ms` (10128), car cette chaîne n’apparaît pas intégralement dans le vocabulaire.
* **“fascinants”** → `Ġfasc` (7751) + `in` (259) + `ants` (1187) : la racine “fasc” est courante, d’où un ID modéré, tandis que “in” et “ants” sont de très petits sous‑mots fréquents ([Hugging Face][2]).
* **“puissants”** → `Ġpu` (47574) + `iss` (747) + `ants` (1187) : “pu” reçoit un ID élevé car ce fragment est rare, illustrant la granularité byte‑level qui découpe même des débuts de mot peu fréquents.

---

### 3. Fréquence vs rareté indiquée par l’ID

* Un **ID faible ou moyen** (p. ex. 259 pour `in`, 747 pour `iss`) indique un sous‑mot fréquent dans le corpus d’entraînement.
* Un **ID très élevé** (p. ex. 47574 pour `Ġpu`) signale un fragment rare, soulignant la granularité fine du tokenizer byte‑level BPE ([Hugging Face][2]).

---

### 4. Implications pratiques

* **Reconstruction de mots** : pour reformer “fascinants”, le modèle assemble trois tokens, ce qui permet de gérer un vocabulaire limité tout en couvrant une large palette de termes ([Hugging Face][3]).
* **Gestion des néologismes** : “LLMs” n’est pas dans le vocabulaire, il est donc décomposé, ce qui permet au modèle de traiter de nouveaux acronymes ou mots inexistants.



## 🌟 Exercise 4 : Pretraining vs. Fine-Tuning
Explain with your own words the two phases of Pretraining and FineTuning in Transformers.

###  1. Pré‑entraînement (Pretraining)

* **But** : construire une **base générale de connaissances linguistiques**, structurales et contextuelles.
* **Comment** : on entraîne le Transformer en auto‑supervision sur **d’énormes quantités de texte non étiqueté** (trillions de tokens). Le modèle apprend à prédire le token suivant ou à inférer des mots manquants, ce qui lui permet de saisir grammaire, vocabulaire, faits, et certains raisonnements ([Reddit][1], [entrypointai.com][2]).
* **Caractéristiques** :

  * très coûteux en données et calcul,
  * ouvre un **large spectre de compétences** transférables à de nombreuses tâches.


---

###  2. Affinage (Fine‑Tuning)

* **But** : spécialiser le modèle sur une **tâche ou un domaine précis** (classification, résumé, Q\&A, style conversationnel…).
* **Comment** : on continue l’entraînement avec un **jeu de données plus petit et étiqueté**, adapté à la tâche. Seules certaines couches ou tous les paramètres peuvent être mis à jour ([Hugging Face Forums][3]).
* **Bienfaits** :

  * **moins coûteux** en ressources que le pré‑entraînement ([llmmodels.org][4]),
  * améliore la **pertinence et la précision** sur des tâches spécifiques,
  * permet d’ajouter des comportements : suite d’instructions (instruct‑tuning), préférences humaines (via RLHF), etc. ([Wikipédia][5])

---

###  Comparaison synthétique

| Phase            | Données                           | Objectif principal                         | Ressources requises |
| ---------------- | --------------------------------- | ------------------------------------------ | ------------------- |
| Pré‑entraînement | Très grandes bases non étiquetées | Apprendre connaissance générale du langage | Très élevé          |
| Fine‑Tuning      | Jeu plus petit, étiqueté          | Adapter à une tâche ou domaine spécifique  | Modéré (plus léger) |

---

###  Pourquoi ces deux étapes ?

* Le **pré‑entraînement** donne un modèle doté d’une **compréhension universelle** du langage, évitant de partir de zéro pour chaque nouvelle tâche ([lightly.ai][6], [Reddit][7]).
* Le **fine‑tuning** tire parti de cette base pour créer des **applications précises et efficaces**, avec moins de données et de cost.

---

###  En résumé

1. **Pré‑entraînement** = fondation large + apprentissage général.
2. **Fine‑tuning** = spécialisation efficace pour des usages concrets.

### 1. Exemple complet de génération de texte

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

# 1. Définir le texte de départ
input_text = "Hier, j'ai découvert un secret incroyable :"

# 2. Charger un modèle léger (GPT‑2)
model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# 3. Créer le pipeline de génération
generator = pipeline("text-generation", model=model, tokenizer=tokenizer)

# 4. Générer du texte
output = generator(input_text, max_length=50, num_return_sequences=1)

# 5. Extraire l’ID et le texte généré
output_ids = output[0]["generated_token_ids"] if "generated_token_ids" in output[0] else None
output_text = output[0]["generated_text"]

# 6. Affichage des résultats
print(f"Input: {input_text}")
print(f"Generated Output: {output_text}")
if output_ids:
    print(f"Output IDs: {output_ids}")
```

Le texte généré par GPT‑2 à partir du prompt **“Hier, j’ai découvert un secret incroyable :”** montre une **répétition systématique** et une **dérive thématique** : après une introduction cohérente, le modèle se lance dans une boucle de phrases redondantes, principalement autour de la « santé » et de la « Commune » (§1). Ce comportement est un cas typique de **degénération textuelle** dû à un **décodage maximaliste** (greedy) et à la **propension des LLMs à renforcer les séquences déjà produites** (§2).

---

### 1. Caractéristiques de la génération

1. **Qualité locale** : la première phrase suit bien la grammaire et le contexte du prompt, avec une cohérence sémantique acceptable ([Wikipédia][1]).
2. **Répétition de séquences** : on observe des répétitions quasi‑identiques de segments comme

   > “et comme cette moi des britiques, et comme cette moi des britiques, …”
   > Cette boucle illustre le **self‑reinforcement effect** : chaque occurrence augmente la probabilité de sa réapparition ([arXiv][2]).
3. **Dérive thématique** : le contenu passe du « secret » initial à un discours sur la **Commune de Paris**, sans transition claire, montrant l’absence de **mémoire de long terme** et la tendance à **strayer off‑topic** pour remplir la longueur demandée ([Wikipédia][1]).

---

### 2. Mécanismes à l’origine des boucles

1. **Décodage greedy** ou mal paramétré : la sélection du token le plus probable à chaque pas renforce les mêmes séquences, conduisant à des boucles infinies ([Medium][3]).
2. **Self‑reinforcement** : les modèles comme GPT‑2 ont un biais naturel à répéter les phrases déjà générées ; plus une phrase apparaît, plus sa probabilité augmente (§1.2) ([arXiv][2]).
3. **Données d’entraînement** : si le corpus contient des répétitions ou des structures similaires, ces motifs se reproduisent à l’infini en génération libre ([Reddit][4]).

---

### 3. Solutions pour atténuer les répétitions

1. **Stratégies de décodage**

   * **Top‑k / Top‑p sampling** : limite la recherche aux K meilleurs candidats ou à une masse de probabilité p, réduisant les choix trop dominants ([MachineLearningMastery.com][5]).
   * **Température > 1** : uniformise la distribution, favorisant la diversité ([MachineLearningMastery.com][5]).
2. **Penalisation des répétitions**

   * **Unlikelihood training** ou méthodes comme **DITTO** : le modèle apprend à pénaliser la génération de séquences déjà vues, diminuant l’effet de boucle ([openreview.net][6]).
   * **Self‑contrastive training** : oppose un checkpoint précoce au modèle courant pour réduire le biais répétitif ([arXiv][7]).
3. **Post‑traitement**

   * **Filtrage heuristique** : détecter et supprimer les répétitions suivantes un certain seuil.
   * **Troncature de la génération** : limiter `max_length` pour éviter les longues digressions (§1.3).

---

### 4. Recommandations pratiques pour le notebook

* **Expérimenter** avec `generator(..., do_sample=True, top_k=50, top_p=0.95, temperature=0.8)` pour améliorer la diversité.
* **Visualiser** les **logits** et la distribution de probabilité des tokens à chaque étape pour comprendre la montée en puissance des boucles ([Medium][3]).
* **Tester** des versions fine‑tunées avec **unlikelihood losses** si tu souhaites approfondir le contrôle de la génération.

```python
#  Visualisations avancées — correction gradients
!pip install bertviz seaborn --quiet

from transformers import AutoTokenizer, AutoModel
import matplotlib.pyplot as plt
import seaborn as sns
import torch
from bertviz import head_view, model_view
from bertviz.transformers_neuron_view import GPT2Model
from bertviz.neuron_view import show as neuron_view

text = "Voici un exemple pour visualiser différents aspects d'un LLM."

tokenizer = AutoTokenizer.from_pretrained("gpt2")
model = AutoModel.from_pretrained("gpt2", output_attentions=True, output_hidden_states=True)
model_neuron = GPT2Model.from_pretrained("gpt2", output_attentions=True, output_hidden_states=True)

inputs = tokenizer(text, return_tensors='pt')
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

outputs = model(**inputs)
attentions = outputs.attentions
hidden_states = outputs.hidden_states

head_view(attentions, tokens)
model_view(attentions, tokens)
neuron_view(model_neuron, "gpt2", tokenizer, text, layer=0, head=0)

# Heatmap attention
attn_matrix = attentions[0][0][0].detach().numpy()
plt.figure(figsize=(6,5))
sns.heatmap(attn_matrix, xticklabels=tokens, yticklabels=tokens, cmap="viridis")
plt.title("Heatmap Attention – Layer 0, Head 0")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

#  Correction : conserver les gradients sur inputs_embeds
emb_layer = model.get_input_embeddings()
inputs_embeds = emb_layer(inputs["input_ids"])
inputs_embeds.retain_grad()  # <-- important

# Passage avec inputs_embeds
outputs2 = model(inputs_embeds=inputs_embeds, output_hidden_states=True)
loss = outputs2.hidden_states[-1].mean()
loss.backward()

# Maintenant les gradients existent
grads = inputs_embeds.grad[0].abs().mean(dim=1).detach().numpy()
plt.figure(figsize=(8,3))
plt.plot(grads, marker='o')
plt.xticks(range(len(tokens)), tokens, rotation=45)
plt.title("Gradient moyen des embeddings")
plt.tight_layout()
plt.show()

# Visualisation des états cachés par couche
means = [hs[0].mean().item() for hs in hidden_states]
plt.figure(figsize=(6,4))
plt.plot(means, marker='o')
plt.xlabel("Couches")
plt.ylabel("Moyenne hidden state")
plt.title("Évolution hidden state par couche")
plt.show()
```

#### 1.1. Représentation

* Les **tokens à gauche** sont ceux qui « attendent » (queries) et ceux à droite sont ceux qui sont « attendus » (keys).
* Chaque **ligne** relie un token query à un token key, et **l’épaisseur** (ou l’intensité) de la ligne reflète la **force** du poids d’attention entre ces deux tokens ([GitHub][1]).

#### 1.2. Sélection interactive

* Le menu “Layer” et “Head” permet de **naviguer** entre les différentes têtes de chaque couche et de filtrer la vue sur un ou plusieurs heads ([Medium][2]).
* Cette exploration dynamique facilite la comparaison des motifs d’attention tête par tête et couche par couche ([ar5iv][3]).

---

### 2. Interprétation du pattern observé

#### 2.1. Structure triangulaire

* La forme **triangulaire** de l’attention (poids concentrés sur la diagonale et les éléments proches de la diagonale) est une **signature du masque causal** utilisé en génération autoregressive – un token ne peut pas voir ceux qui viennent **après** lui ([aclanthology.org][4]).

#### 2.2. Biais positionnel local

* En première couche, les têtes sont souvent **spécialisées** pour capturer la **proximité** (contexte local) plutôt que des dépendances à longue portée ([aclanthology.org][4]).
* Cela correspond au **“next-word attention pattern”** où chaque mot se tourne surtout vers ses prédécesseurs immédiats, un peu comme un modèle n‑gram local ([Medium][5]).

#### 2.3. Stabilisation du contexte initial

* Le pic d’attention particulièrement fort sur le **premier token** (« Vo ») reflète un effet de **stabilisation du contexte** : ce token sert de point d’ancrage, surtout dans les têtes de couche 0, où on jette les bases du contexte global ([Debugging Machine Learning Models][6]).

---

### 3. Importance pédagogique

1. **Compréhension des patterns basiques** : avant d’étudier des dépendances lexicales ou syntaxiques complexes, ces têtes de couche 0 agissent comme des **filtrages positionnels**, garantissant que chaque token incorpore d’abord son propre embedding et ceux de ses proches voisins ([ar5iv][7]).
2. **Diagnostic de l’architecture** : en confirmant que les têtes initiales se concentrent sur le contexte local, on valide le bon **fonctionnement du masque causal** et on peut ensuite cibler les couches intermédiaires pour analyser les patrons plus abstraits (entités nommées, relations à distance, etc.) ([aclanthology.org][4]).
3. **Réglage de la granularité** : comprendre que Layer 0 privilégie le local permet d’ajuster finement les expérimentations (p. ex. en binarisant les têtes pour forcer plus de focus thématique dans les couches supérieures) ([ar5iv][3]).

---

### 4. Perspectives pour la suite

* **Explorer d’autres heads** dans Layer 0 pour repérer des patterns complémentaires (p. ex. “bag-of-words” ou “delimiter-focused”).
* **Comparer** avec des têtes en couches plus profondes (Layers 9–11) pour observer l’évolution de la granularité et la naissance de **patterns sémantiques** (p. ex. focus sur des entités ou relations sujet‑verbe) ([Debugging Machine Learning Models][6]).
* **Intégrer** les vues **model\_view** et **neuron\_view** pour un diagnostic complet des mécanismes query/key et de leurs neuron activations, afin d’aligner l’interprétabilité sur les besoins de l’application finale ([arXiv][8]).

## 1. Heatmap Attention – Layer 0, Head 0

* On observe une **valeur élevée sur la diagonale** indiquant que chaque token s’attend surtout à lui‑même ou au token précédent, typique d’une tête causale en première couche ([aclanthology.org][1]).
* Le **fort pic** sur le premier token (“Vo”) montre que cette tête accorde une priorité majeure au **début de la séquence**, comme souvent en couche 0 où l’on consolide le contexte local ([ResearchGate][2]).
* La **forme triangulaire** (diminution progressive de l’attention pour les tokens lointains) reflète le **masque causal** et un noyau positionnel simple ([ewadirect.com][3]).
* Ce comportement correspond aux **“positional kernels”** décrits pour les premiers heads, qui capturent la structure positionnelle avant d’intégrer le contenu sémantique ([lesswrong.com][4]).

---

## 2. Gradient moyen des embeddings

* Le token **“ici”** (deuxième position) présente le **gradient moyen le plus élevé**, indiquant qu’il contribue fortement à la fonction de perte moyenne et donc qu’il est jugé **informatif** lors de la rétropropagation ([arXiv][5]).
* Les tokens plus **rares** ou plus **spécifiques** (e.g., “Gdíff”) génèrent souvent des **gradients plus importants**, car le modèle ajuste davantage leurs embeddings pour améliorer la prédiction ([arXiv][6]).
* La **diminution graduelle** des gradients pour les tokens ultérieurs montre une **influence décroissante** des mots éloignés du début, typique d’une accumulation contextuelle en cascade ([docs.seldon.io][7]).

---

## 3. Évolution des états cachés par couche

* La **norme moyenne** des hidden states **augmente linéairement** de la couche 0 à la couche 11, reflétant la **richesse croissante** des représentations sémantiques à chaque transformation ([arXiv][8]).
* Le **pic** à la couche 11 puis la **légère baisse** en couche 12 suggèrent une **compression finale** de l’information avant la sortie, où les états sont consolidés pour la prédiction de tokens ([arXiv][9]).
* Cette dynamique est cohérente avec les analyses de **GPT‑2 small**, où les couches intermédiaires construisent progressivement le contexte, et la couche finale affine et sélectionne les caractéristiques clés ([catherinesyeh.github.io][10]).
